```python
cards = ('heart','spade','diamond','club')
```

```python
print('heart' in cards)
```

```python
print('spade' not in cards)
```

```python
for card in cards:
    print(card.upper())
```

```python
l_cards.append('random')
```

```python
t1 = (1,2,3)
t2 = (10,20,30)
t3 = ('one','two','three')
```

```python
n1,n2,n3 = t1 #tuple unpacking
print(n2)
print(n1)
print(n3)
```

```python
#zipping
zipped = tuple(zip(t1,t2,t3))
# when t1 = (1,2,3)
# t2 = (10,20,30)
# t3 = ('one','two')
# zipped = ((1, 10, 'one'), (2, 20, 'two'))
```
