```python
capitals = {
    'India':'New Delhi',
    'US':'Washington DC',
    'France':'Paris',
    'UK':'London'
}
```

```python
capitals
```

```python
for key in capitals.keys():
    print(f'{key} -- {capitals[key]}')
    
```

```python
for val in capitals.values():
    print(val)
    
```

```python
for key, val in capitals.items():
    print(f'{key} -- {val}')
    
```
