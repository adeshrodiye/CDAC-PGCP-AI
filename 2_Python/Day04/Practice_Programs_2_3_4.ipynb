Q1. Define a function overlapping () that takes two lists and returns True if they have at
least one member in common, False otherwise.

```python
def overlapping(list1, list2):
    for item1 in list1:
        for item2 in list2:
            if item1 == item2:
                return True
    return False
overlapping([1,5,6,7,9],[4,2,3,6,8])
```

Q.2.In English, present participle is formed by adding suffix -ing to infinite form: go -> going. A simple set of rules can be given as follows:
 a. If the verb ends in e, drop the e and add ing 
 b. If the verb ends in ie, change ie to y and add ing
Write a function make_ing_form() which accepts a list of verbs and returns a dictionary with verb : present participle


```python
def make_ing_form(verbs):
    result = {}
    for verb in verbs:
        if verb.endswith("ie"):
            ing_form = verb[:-2] + "ying"
        elif verb.endswith("e"):
            ing_form = verb[:-1] + "ing"
        else:
            ing_form = verb + "ing"

        result[verb] = ing_form
        
    return result

make_ing_form(['lie','go','make','play'])
```

Q.3. Function display_greeting(message) prints message sent as argument as today's greeting. Decorate the function using appropriate decorated so that the greeting is displayed using Uppercase.

```python
def uppercase_decorator(function):
    def wrapper(message):
        upper_message = message.upper()
        return function(upper_message)
    return wrapper

@uppercase_decorator
def display_greeting(message):
    print(message)

display_greeting("good morning. have a great day!")
```

Q. 4. Create series of 'n' prime numbers and display first 10

```python
def prime_generator():
    
    num = 2
    while True:
        is_prime = True
        for i in range(2, (num//2)+1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            yield num

        num += 1

prime = prime_generator()

n = 10
print("First 10 prime Numbers")
for i in range(n):
    print(next(prime))
        
```

Q.6
Following data displays min/max/average temp for cities
weather= [{'Mumbai' : [28, 30, 32]},.....]

1. Print the weather data
2. Print the city with maximum/min temp
3. Print all the cities that expereince min temp more than 30 degree
4. Create a dictionary to print 'City':'Ave temp'


```python
# 1.Print the weather data
weather = [
    {'Mumbai': [28, 35, 31]},
    {'Delhi': [20, 42, 31]},
    {'Pune': [18, 28, 23]},
    {'Chennai': [31, 38, 34]}
]

print("Weather Data:")
for data in weather:
    for city,temps in data.items():
         print(f"{city}: Min = {temps[0]}°C, Max = {temps[1]}°C, Avg = {temps[2]}°C")
```

```python
# 2. Print the city with maximum/min temp
max_temp_city = max(weather, key=lambda t: list(t.values())[0][1])
min_temp_city = min(weather, key=lambda t: list(t.values())[0][0])

max_city_name, max_values = list(max_temp_city.items())[0]
min_city_name, min_values = list(min_temp_city.items())[0]

print(f"City with highest MAX temp: {max_city_name} ({max_values[1]}°C)")
print(f"City with lowest MIN temp:  {min_city_name} ({min_values[0]}°C)")
```

```python
# 3. Print all the cities that experience min temp more than 30 degrees
print("3. Cities with min temp > 30°C:")
for data in weather:
    for city, temps in data.items():
        if temps[0] > 30:
            print(city)
```

```python
# 4. Create a dictionary to print 'City':'Avg temp'
avg_temp_dict = {
    city: temps[2] 
    for city_dict in weather 
    for city, temps in city_dict.items()
}

print("4. City to Average Temp Dictionary:")
print(f"   {avg_temp_dict}")
```
