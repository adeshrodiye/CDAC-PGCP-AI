```python
countries = df.groupby('Country')
countries
```

```python
india = countries.get_group('India')
india.
```

```python
df['ConvertedCompYearly'].agg(['min','mean','max'])
```

```python
countries = df.groupby('Country')
countries
```

```python
countries
```

```python
countries['YearsCode'].value_counts()
```

```python
countries['ConvertedCompYearly'].agg(['min','mean','max'])
```

```python
countries.get_group('India')['ConvertedCompYearly'].agg(['min','mean','max'])
```
