# Line Chart

```python
import pandas as pd
import matplotlib.pyplot as plt
#from matplotlib import pyplot as plt
#file handling
#regular expression
```

```python
df = pd.read_csv("Population.csv")
df
```

```python
df.drop(index=[14,15,16],inplace=True)
```

```python
df
```

```python
plt.plot(df["Year"],df["Population"]) 
plt.plot(df["Year"],df["Urban Population"]) 
```

```python
df.isna()
```

```python
plt.plot(df["Year"],df["Population"],ls='--',lw=0.8,color="r", label="Total Population")
plt.plot(df["Year"],df["Urban Population"],ls='-',lw=0.8,color="b", label="Urban Population")
plt.xlabel("Year")
plt.ylabel("Population in Billions")
plt.title("Population of India Over the Years")
plt.legend()
plt.show()
```
