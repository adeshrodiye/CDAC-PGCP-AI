# Input Output 

```python
name = input("Enter Your Name:")
age = int(input("Enter Your Age:"))
print("YOUR ENTERED NAME : ", name, "and AGE :",age)
```

```python
#format String
print("You Entered Name: {} and Age {}".format(name,age))
```

```python
#format string operator
data = 4e+6
print(data)
```

```python
print(f'{data:.2f}')
```

```python
data = 4e-6
print(data)
print(f'{data:.6f}')
```

```python
data = 1000000
print(data)
print(f'{data:,}')
print(f'{data:.2f}')
```

```python
data = 0.95
print(data)
print(f'{data:%}')
```

```python
data = 55
print(f'{data:05d}')
```

```python
name = "Adesh"
print(f'{name:<10}')
print(f'{name:>10}')
print(f'{name:^100}')
```
