```python
import Math
# from Math import add,subtract,divide,multiply
# from Math import multiply as mult
```

```python

while True:

    choice = int(input("\nEnter Your Choice:\n1.Addition\n2.Subtraction\n3.Multiplication\n4.Division\n5.End\n"))

    if 0 < choice < 5:
        a = int(input("Enter 1st number:\n"))
        b = int(input("Enter 2nd number:\n"))
        
    match choice:
        case 1:
            Math.add(a,b)
        case 2:
            Math.subtract(a,b)
        case 3:
            Math.multiply(a,b)
        case 4:
            Math.divide(a,b)
        case 5:
            print("Exit")
            break
        case _:
            print("Invalid Input")
    
```
