# Generators

- speacial type of function use to return series of value
- generator functions use *yield* to produce a series of results over time
- When yield is executed, it pauses the function, returns the current value and retains the state of the function
- This allows the function to continue from same point when called again

```python
def get_number():
    i = 0
    while True:
        yield i
        i +=1

series = get_number()
for i in  range(100):
    print(next(series))
```

```python
def get_fib(end):
    i = 0
    a,b = 0,1
    while i < end:
        yield b
        a, b = b, a+b
        i+=1
fib_series = get_fib(20)
for i in range(20):
    next(fib_series)
```
