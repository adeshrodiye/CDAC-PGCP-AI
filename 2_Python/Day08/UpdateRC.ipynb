```python
import pandas as pd
```

```python
data = {'FullName' : ['Rucha N','Rohit K','Kedar P','Ankita G'],
        'Email' : ['rucha07@gmail.com','rohit,10@gmail.com','kedartennis@gmail.com','ankita13@gmail.com'],
        'Salary' : [100000,120000,90000,75000]}

df = pd.DataFrame(data)
df
```

```python
df['Bonus'] = df['Salary']*0.15
df
```

```python
df['Gender'] = ['F','M','M','F']
df
```

```python
df.columns = ["Full Name","Email id","Salary","Bonus","Gender"]
df
```

```python
df.columns = [s.lower() for s in df.columns]
df
```

```python
df.columns = list(map(lambda x: x.title(),df.columns))
df
```

```python
df.columns = list(map(lambda x:x.upper(),df.columns))
df
```
