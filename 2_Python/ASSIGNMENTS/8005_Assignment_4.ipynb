## Q1. Create a DataFrame containing student information like Name, Age, Department, %Marks

a.	Display column names of the DF

b.	Show min/max and average marks

c.	Get unique values and student count for each Department

```python
import pandas as pd
```

```python
df = pd.DataFrame({"Name":['Adesh','Sarthak','Vivek','Raghav','Vishal','Yugal'],
                   "Age":[22,23,21,23,24,20],
                   "Department":["AIDS","CS","ML","AIDS","IT","CS"],
                   "%Marks":[92,98,92,95,90,93]})
df
```

### A.  Display column names of the DF

```python
df.columns
```

### B. Show min/max and average marks

```python
df["%Marks"].agg(['min','max','average'])
```

###  C. Get unique values and student count for each Department

```python
dept = df["Department"].value_counts() # IT RETURNS SERIES OF STUDENT COUNT FOR EACH DEPARTMENT
dept
```

```python
# GROUP BY DEPARTMENT,.size()- Count number of students in each department and return Series, .
dept = df.groupby("Department").size().reset_index(name="Student Count")
dept
```

```python
filt = solar["Installed Capacity (MW)"]>500
solar[filt]
```

```python
# Standard query: 
solar.query("`Installed Capacity (MW)` > 500") #
```

```python
# using Standard Boolean Indexing:
solar[solar["Installed Capacity (MW)"] > 500]
```

### B. Display plant details for New York plant

```python
solar
```

```python
solar[solar['State']=='New York']
```

### C. For all the plants display Average MW Per Plant and Generation (GWh)

```python
solar[['State','Average MW Per Plant','Generation (GWh)']]
```

### D. Sort the plants by Generation (GWh) (ascending) and Installed Capacity (MW)(descending)

```python
solar.sort_values(by=['Generation (GWh)','Installed Capacity (MW)'],ascending=[True,False])
```

### E. List top 5 plants for their Generation

```python
sorted = solar.sort_values(by="Generation (GWh)",ascending=False)
sorted.head(5)
```

```python
#using nlragest
 solar.nlargest(5, "Generation (GWh)")
```

### F. Display all plant details for states – California, Nevada, Arizona and Texas

```python
solar.query("State in ['California', 'Nevada', 'Arizona', 'Texas']")
```

### G. Group the plants by region, find – min/max capacity plant for each group

```python
solar.groupby("Region")['Installed Capacity (MW)'].agg(["min","max"])
```

```python
#assign names to new column 
solar.groupby("Region").agg(
    Min_Capacity=("Installed Capacity (MW)", "min"),
    Max_Capacity=("Installed Capacity (MW)", "max"),
)
```
