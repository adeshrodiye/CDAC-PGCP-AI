### 1. Draw a bar chart showing GDP percentage for European countires. [use gdpminder dataset] Use different semantics as orientation, bar mode, bar norm, text to improve your chart For the same dataset show GDP over the years using animation frame as Country

```python
dir(px.data)
```

```python
df = px.data.gapminder()
df
```

```python
df['Total_gdp'] = df['gdpPercap'] * df['pop']
df
```

```python
df['continent'].unique()
```

```python
df1 = df.loc[df['continent']=='Europe']
df1
```

```python
#calculate %
df1['gdp_percent'] = df1.groupby('year')['Total_gdp'].transform(lambda x: (x / x.sum())*100)
df1
```

```python

fig = px.bar(df1,x="gdp_percent",y="year",color="country",orientation="h",barmode="stack",text_auto='.2f',
             title = "GDP percentage for European countires",labels={"total_gdp" : "Gdp_ Share" , "year" : "Year", "country" : "Country"})
fig
```

```python
fig = px.bar(df1,x="gdp_percent",y="year",color="country",animation_frame = 'year',orientation="h",barmode="stack",text_auto='.2f',
             title = "GDP percentage for European countires",labels={"total_gdp" : "Gdp_ Share" , "year" : "Year", "country" : "Country"})
fig
```
