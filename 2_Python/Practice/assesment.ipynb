```python
marks = [10,12,1,3,7]
n = len(marks)
```

```python
for i in range(n):
    for j in range(0,n-1-i):
        if marks[j] > marks[j+1]:
            marks[j],marks[j+1]=marks[j+1],marks[j]
```

```python
min = sum(marks[:3])
max = sum(marks[-3:])
print(min)
print(max)
```

```python
n = int(input("Enter number of games:"))

scores = []

for i in range(n):
    score = int(input(f"Enter score of game {i+1} : "))
    scores.append(score)

scores
```

```python
min_score = scores[0]
max_score = scores[0]

highest_breaks = 0
lowest_breaks = 0

for score in scores[1:]:
    if score > max_score:
        max_score = score
        highest_breaks += 1
    elif score < min_score:
        min_score = score
        lowest_breaks += 1

output = [highest_breaks,lowest_breaks]
output
```

```python
n = int(input("Enter size of list:"))

list1 = []

for i in range(n):
    l = int(input(f"Enter number {i+1} : "))
    list1.append(l)

list1
```

```python
from collections import Counter
reminder = [x % 5 for x in list1]
print("Reminders",reminder)
reminders_count = Counter(reminder)
print("reminders_count",reminders_count)
```

```python
total_strength = sum(c*(c-1) for c in reminders_count.values())
total_strength
```

```python
def evenFactors(n):
    print("Even Factors of n are : ")
    for i in range(1,n+1):
        if( n%i==0 and i%2==0):
            print(i," ")
evenFactors(60)
```
