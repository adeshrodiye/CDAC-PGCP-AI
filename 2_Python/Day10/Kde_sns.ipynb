# KDE - Kernel Density Estimate

```python
import seaborn as sns
```

```python
df = sns.load_dataset("tips")
```

```python
df
```

```python
sns.kdeplot(df,x="total_bill")
```

```python
sns.kdeplot(df,x="total_bill", hue="sex", bw_adjust=0.25)
```

```python
sns.kdeplot(df,x="total_bill", hue="sex", fill=True)
```

```python
sns.kdeplot(df,x="total_bill", hue="sex", fill=True)
```
