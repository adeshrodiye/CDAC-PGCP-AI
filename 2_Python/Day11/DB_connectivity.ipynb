```python
import sqlite3

def get_connection(path):
    try:
        conn = sqlite3.connect(path)
        print("Connection Successful !")
        return conn
    except sqlite3.Error as er:
        print(er)
        return None

def execute_query(conn, query):
    try:
        cursor = conn.cursor()
        cursor.execute(query)
        conn.commit()
        print("Query Executed Successfully !")
    except sqlite3.Error as er:
        print(er)

def execute_read_query(conn, query):
    try:
        cursor = conn.cursor()  # Added () here
        cursor.execute(query)
        d_result = cursor.fetchall()
        return d_result
    except sqlite3.Error as er:  # Changed to sqlite3.Error
        print(er)
        return None
```

```python
create_table = """CREATE TABLE IF NOT EXISTS users
(id INTEGER PRIMARY KEY AUTOINCREMENT,
  name TEXT NOT NULL,
  age INTEGER,
  gender TEXT,
  nationality TEXT);"""

add_users = """
INSERT INTO
  users (name, age, gender, nationality)
VALUES
  ('James', 25, 'male', 'USA'),
  ('Leila', 32, 'female', 'France'),
  ('Brigitte', 35, 'female', 'England'),
  ('Mike', 40, 'male', 'Denmark'),
  ('Elizabeth', 21, 'female', 'Canada');
"""
fetch_users = """SELECT * from users;"""

update_user = """
UPDATE 
    users
SET
    age = 22
WHERE 
    name = 'Mike'
"""

delete_user = """
DELETE from users WHERE id = 5"""

select_females = """SELECT name, age, nationality
FROM users WHERE gender = 'female'"""
```

```python
path = "users.sqlite3"
conn = get_connection(path)
```

```python
execute_query(conn,create_table)
```

```python
execute_query(conn,add_users)
```

```python
select_all = """Select *From users"""
```

```python
results = execute_read_query(conn,fetch_users)
```

```python
for it in results:
    print(it)
```

```python
execute_query(conn,update_user)
```

```python
results = execute_read_query(conn,select_all)
for q_results in results:
    print(q_results)
```

```python
results = execute_read_query(conn,select_females)
for i in results:
    print(i)
```

```python
execute_query(conn,delete_user)
results = execute_read_query(conn, select_all)
for i in results:
    print(i)
```

```python
new_path = "books.sqlite3"
conn = get_connection(new_path)
```

```python
# Book - isbn, title, author, price
create_table = """CREATE TABLE IF NOT EXISTS books
(isbn INTEGER PRIMARY KEY AUTOINCREMENT,
  title TEXT NOT NULL,
  author TEXT,
  publication TEXT,
  price INTEGER);"""

execute_query(conn,create_table)
```

```python
select_all = """Select *From books"""
results = execute_read_query(conn,select_all)
for i in results:
    print(i)
```
