# ONE HOT ENCODING

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
df = pd.read_csv("titanic (1).csv")
df.head()
```

```python
selctef_col = ["survived","pclass","gender","age","sibsp","parch","embarked"]
df = df[selctef_col]
df
```

```python
X = df.drop("survived",axis=1)
y = df["survived"]
```

```python
# One hot Encoding categorical features

categorical_features = ["pclass","gender","sibsp","parch","embarked"]

encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

```

```python
encoded_data = encoder.fit_transform(df[categorical_features])
encoded_data
```

```python
feature_names = encoder.get_feature_names_out(categorical_features)
```

```python
feature_names
```

```python
encoded_df = pd.DataFrame(encoded_data,columns=feature_names)
encoded_df
```

```python
missing_values_age = X["age"].isna().any()
```

```python
# scale the numeric age col: handle missing values in age before scaling
if missing_values_age:
    print("Hanling nulls ...")
    imputer = SimpleImputer(strategy="mean")
    imputer.fit(X[["age"]])
    X["age"] = imputer.transform(X[["age"]])[:,0]
print(X.age)
```

```python
scaler = StandardScaler()
scaler.fit(X[["age"]])
X_scaled = X.copy()
```

```python
X_scaled
```

```python
X_scaled.loc[:,"age"] = scaler.transform(X[["age"]])
X_scaled


```

```python
# Apend the scaled age to the original encoded df dataframe
df_merged = pd.concat([encoded_df, X_scaled["age"]],axis=1)
df_merged.head()
```
