# Subplots

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df
```

```python
plt.figure(figsize=(12,5))
 # subplot 1
avg_tip_day = df.groupby('day')['tip'].mean()
plt.subplot(1,2,1)
plt.bar(avg_tip_day.index,avg_tip_day.values,color = ['lightblue','lightgreen','pink','orange'])
plt.title('Avg. Tip per day')
plt.ylabel('Avg TIp')

#subplot 2
smoker_count = df['smoker'].value_counts()
plt.subplot(1,2,2)
plt.pie(smoker_count,labels=smoker_count.index,autopct='%1.1f%%',colors=['gold','lightcoral'])
plt.title('Smoker vs non-Smoker')
plt.tight_layout()
plt.show()
```

```python
# Object oriented method  
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1
axes[0].plot(df['total_bill'].iloc[:20], marker='o', linestyle='--', color='r')
axes[0].set_title('Total Bill')
axes[0].set_xlabel('Record Index')
axes[0].set_ylabel('Total Bill')

# Plot 2
ave_tip_gender = df.groupby("sex", observed=False)["tip"].mean()
axes[1].barh(ave_tip_gender.index, ave_tip_gender.values, color=['pink', 'lightblue'])
axes[1].set_title('Average Tip by Gender')
axes[1].set_xlabel('Average Tip')

plt.tight_layout()
plt.show()
```
