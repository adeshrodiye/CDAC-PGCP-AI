```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('tips.csv')
df
```

# Matplot 

### Histogram

```python
plt.figure(figsize=(6,4)) ## size of graph
plt.hist(df['total_bill'],bins = 10,color = 'brown',edgecolor = 'gold')
plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

### Bar plot

#### average tip by gender

```python
avg_tip_gender = df.groupby('gender')['tip'].mean()
plt.figure(figsize=(6,4))
plt.bar(avg_tip_gender.index,avg_tip_gender.values,color = ['pink','lightblue'])
plt.title('Average Tip by gender')
plt.ylabel('Average Tip')
plt.show()
```

### line chart

```python
plt.figure(figsize=(13,5))
plt.plot(df.index,df['total_bill'],marker = '*',linestyle = '-',color='purple')
plt.title('Total Bill across Records')
plt.xlabel('Index')
plt.ylabel('Total Bill')
plt.show()
```

### Scatter plot

```python
plt.figure(figsize=(6,4))
plt.scatter(df['total_bill'],df['tip'],color = 'orange',edgecolor = 'black')
plt.title('Total Bill vs. Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

# Seaborn

### Histogram

```python
sns.histplot(df['total_bill'],bins=10,color = 'darkgreen')
plt.title('Distribution of Total Bill')
plt.show()
```

```python
sns.boxplot(x = 'gender',y='tip',data = df,hue = 'gender')
```

### Count plot

```python
sns.countplot(x = 'smoker',data=df,hue='smoker')
```

### Scatter plot

```python
sns.regplot(x = 'total_bill',y='tip',data = df,color = 'black')
```

### Bar plot

```python
sns.barplot( x = 'day',y = 'tip',data = df)
```

## Subplots

```python
plt.figure(figsize=(10,8))
plt.subplot(2,2,1)
sns.histplot(df['total_bill'],bins=10,color = 'darkgreen')
plt.subplot(2,2,2)
sns.boxplot(x = 'gender',y='tip',data = df,hue = 'gender')
plt.subplot(2,2,3)
sns.countplot(x = 'smoker',data=df,hue='smoker')
plt.subplot(2,2,4)
sns.regplot(x = 'total_bill',y='tip',data = df,color = 'black')
plt.tight_layout()
plt.show()
```
