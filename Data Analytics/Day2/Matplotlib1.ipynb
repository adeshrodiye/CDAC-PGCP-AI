```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df = sns.load_dataset('tips')
df
```

```python
plt.figure(figsize=(5,4))
plt.hist(df['total_bill'],bins=10,color='purple',edgecolor='black')
plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show();
```

```python
# Hist
plt.figure(figsize=(5,4))
sns.set_style(style="darkgrid")
plt.hist(df['total_bill'],bins=5,color='black',edgecolor='white')
plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

```python
# Bar
avg_tip_gender = df.groupby('sex')['tip'].mean()
plt.figure(figsize=(6,4))
sns.set_style(style="white")
plt.bar(avg_tip_gender.index,avg_tip_gender.values,color=['pink','lightblue'])
plt.title('Average tip by gender')
plt.ylabel('Average Tip')
plt.show()
```

```python
# Hist
plt.figure(figsize=(15,6))
sns.set_style(style="darkgrid")
plt.plot(df.index,df['total_bill'],'r-',marker='o')
plt.title('Total Bill Across Records')
plt.xlabel('Index')
plt.ylabel('Total Bill')
plt.show()
```

```python
# Hist
plt.figure(figsize=(10,8))
sns.set_style(style="darkgrid")
plt.scatter(df['total_bill'],df['tip'],c='b' ,edgecolor = 'black')
plt.title('Total Bill Vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```python
# Hist
plt.figure(figsize=(10,8))
sns.set_style(style="darkgrid")
plt.scatter(df['total_bill'],df['tip'],c='#8ffe09' ,edgecolor = 'black')
plt.title('Total Bill Vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```
