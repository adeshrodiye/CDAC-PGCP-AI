```python
a = np.array([10,20,30])
b = np.array([1,2,3])
a+b
```

```python
a = pd.Series([10,20,30],index=['alice','bob','charlie'])
b = pd.Series([1,2,3,4], index=['bob','charlie','alice','deve'])
print(a+b) # matched by label, not by positions
```
