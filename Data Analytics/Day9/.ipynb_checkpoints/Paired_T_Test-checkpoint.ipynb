# Paired T Test

```python
import pandas as pd
from scipy.stats import ttest_rel
```

```python
df = pd.read_csv("cust_seg.csv")
df.head()
```

```python
usage_data = df[["pre_usage", "Post_usage_1month"]]
usage_data = usage_data.sample(n=30)
```

```python
t_stat, p_value = ttest_rel(usage_data["pre_usage"],usage_data["Post_usage_1month"])
print("t_stat:",t_stat)
print("p_value:",p_value)
```

```python
alpha = 0.05
if p_value < alpha:
    print("Reject null Hypothesis")
else:
    print("Fail to reject null Hypothesis")
```

## Exercise : check whether the urgent threatment is effective (consider t1 adn t2 from hypothermia dataset

```python
df = pd.read_csv("Hypothermia.csv")
df.head()
```

```python
temp_data = df[["t.1", "t.2"]]
temp_data = df.sample(n=30, random_state=42)
```

```python
t_stat, p_value = ttest_rel(temp_data["t.1"],temp_data["t.2"])
print("t_stat:",t_stat)
print("p_value:",p_value)
```

```python
alpha = 0.05
if p_value < alpha:
    print("Reject null Hypothesis : The treatment is effective")
else:
    print("Fail to reject null Hypothesis : No significant temperature increase detected")
```
