```python
from scipy.stats import f_oneway
import pandas as pd
```

```python
df = pd.read_csv("StudentsPerformance.csv")
df.head()
```

```python
df["parental level of education"].unique()
```

```python
g1 = df["math score"][df["parental level of education"] == "some high school"]

g2 = df["math score"][df["parental level of education"] == "high school"]

g3 = df["math score"][df["parental level of education"] == "some college"]

g4 = df["math score"][df["parental level of education"] == "associate's degree"]

g5 = df["math score"][df["parental level of education"] == "bachelor's degree"]

g6 = df["math score"][df["parental level of education"] == "master's degree"]
```

```python
fstat, pval = f_oneway(g1, g2, g3, g4, g5, g6)
print("fstat : ",fstat)
print("pval : ",pval)
alpha = 0.05

if pval < alpha:
    print("Reject H0")
else:
    print("Do not reject H0")
```

## Exercise
### Is there a statistically siginificant differnece between how people tip,whose total bill amounts are <5,5-10,10-15,>15 in the tips dataset(Handel boundary conditions approprriately)

```python
df = pd.read_csv("tips.csv")
df.head()
```

```python
g1 = df["tip"][df["total_bill"] < 5]

g2 = df["tip"][(df["total_bill"] >= 5) & (df["total_bill"] <= 10)]

g3 = df["tip"][(df["total_bill"] > 10) & (df["total_bill"] <= 15)]

g4 = df["tip"][df["total_bill"] > 15]
```

```python
fstat, pval = f_oneway(g1, g2, g3, g4)
print("fstat : ",fstat)
print("pval : ",pval)
alpha = 0.05

if pval < alpha:
    print("Reject H0")
else:
    print("Do not reject H0")
```
