```python
import pandas as pd
from scipy import stats
```

```python
df = pd.read_csv("tips.csv")
df.head()
```

```python
male_tips = df[df['gender'] == 'Male']['tip']
female_tips = df[df['gender'] == 'Female']['tip']
```

```python
male_tips
```

```python
female_tips
```

```python
print("Male Mean tip : ", male_tips.mean())
print("Feale Mean tip : ", female_tips.mean())
```

### Two-sample t-test

```python
t_statistic, p_value = stats.ttest_ind(male_tips,female_tips)
print("T-statistic:",t_statistic)
print("P_value (one-tailed) : ", p_value)
```

```python
alpha = 0.05

if p_value < alpha :
    print("There is significant difference in spending score between Male and Female Group.")
else:
    print("There is no significant difference in spending score between Male and Female Group.")
```

## Ex: is there a statistically significant diff in the average runs runs scored by sachin tendulkar when playing against australia vs pakistan

```python
df = pd.read_csv("Sachin_Tendulkar (1).csv")
df.head()
```

```python
df["Versus"].unique()
```

```python
against_pakistan = df[df["Versus"] == "Pakistan"]["Runs"]
against_australia = df[df["Versus"] == "Australia"]["Runs"]
```

```python
against_pakistan
```

```python
against_australia
```

```python
print("Pakistan Mean Runs : ", against_pakistan.mean())
print("Australia Mean Runs : ", against_australia.mean())
```

```python
t_statistic, p_value = stats.ttest_ind(against_pakistan,against_australia)
print("T-statistic:",t_statistic)
print("P_value  : ", p_value)
```

```python
alpha = 0.05
if p_value < alpha :
    print("There is significant difference in the average runs runs scored by sachin tendulkar when playing against australia vs pakistan")
else:
    print("There is no significant difference in the average runs runs scored by sachin tendulkar when playing against australia vs pakistan")
```
