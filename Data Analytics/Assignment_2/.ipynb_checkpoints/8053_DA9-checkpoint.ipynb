## Q1. mtcars Dataset Analysis

### 1.1 Independent Two-Sample t-Test
- **Hypotheses**:
  - $H_0$: There is no difference in the mean `mpg` between 4-cylinder and 6-cylinder cars ($\mu_4 = \mu_6$).
  - $H_1$: There is a significant difference in the mean `mpg` between 4-cylinder and 6-cylinder cars ($\mu_4 \neq \mu_6$).

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from statsmodels.formula.api import ols
```

```python
df_mtcars = pd.read_csv('/content/mtcars.csv')

mpg_4 = df_mtcars[df_mtcars['cyl'] == 4]['mpg']
mpg_6 = df_mtcars[df_mtcars['cyl'] == 6]['mpg']

print(f"Sample Mean mpg (4-cylinder): {mpg_4.mean():.2f}")
print(f"Sample Mean mpg (6-cylinder): {mpg_6.mean():.2f}")
```

```python
# Two-Sample t-Test
t_stat, p_val = stats.ttest_ind(mpg_4, mpg_6, equal_var=False)

mean_diff = mpg_4.mean() - mpg_6.mean()
n1, n2 = len(mpg_4), len(mpg_6)
se = np.sqrt(mpg_4.var()/n1 + mpg_6.var()/n2)
df_deg = n1 + n2 - 2
ci = stats.t.interval(0.95, df_deg, loc=mean_diff, scale=se)

print(f"t-statistic: {t_stat:.4f}")
print(f"p-value: {p_val:.4e}")
print(f"95% Confidence Interval for the difference: {ci}")
```

#### • Interpretation of mtcars t-Test : If the p-value is $< 0.05$, we reject the null hypothesis and conclude that there is a statistically significant difference in mean `mpg` between 4-cylinder and 6-cylinder cars.

### 1.2 One-Way ANOVA Test (mtcars)
- **Hypotheses**:
  - $H_0$: The mean horsepower (`hp`) is the same across all `gear` groups ($\mu_{gear3} = \mu_{gear4} = \mu_{gear5}$).
  - $H_1$: At least one `gear` group has a different mean horsepower.

```python
# ANOVA
model_mtcars = ols('hp ~ C(gear)', data=df_mtcars).fit()
anova_table = sm.stats.anova_lm(model_mtcars, typ=2)
display(anova_table)
```

# Q2 : Wine Quality Dataset Analysis

### 2.1 Correlation, Covariance and Outlier Analysis

```python
df_wine = pd.read_csv('/content/wine_quality_red.csv', sep = ";")
df_wine.head()
```

```python
# Covariance and Correlation
cov_alc_sulph = df_wine['alcohol'].cov(df_wine['sulphates'])
corr_alc_qual = df_wine['alcohol'].corr(df_wine['quality'])

print(f"Covariance (alcohol vs sulphates): {cov_alc_sulph:.4f}")
print(f"Correlation (alcohol vs quality): {corr_alc_qual:.4f}")
```

```python
# Scatter Plot
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_wine, x='alcohol', y='quality', alpha=0.5)
plt.title('Scatter plot: Alcohol vs Quality')
plt.grid(True)
plt.show()
```

```python
# Outlier detection in alcohol
q1 = df_wine['alcohol'].quantile(0.25)
q3 = df_wine['alcohol'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df_wine[(df_wine['alcohol'] < lower_bound) | (df_wine['alcohol'] > upper_bound)]
df_wine_clean = df_wine[(df_wine['alcohol'] >= lower_bound) & (df_wine['alcohol'] <= upper_bound)]

corr_alc_qual_clean = df_wine_clean['alcohol'].corr(df_wine_clean['quality'])

print(f"Outliers detected in alcohol: {len(outliers)}")
print(f"Updated Correlation (alcohol vs quality) without outliers: {corr_alc_qual_clean:.4f}")
```

### 2 Independent Two-Sample t-Test (Wine Quality)
- **Groups**:
  - Group A: Quality $\le$ 5
  - Group B: Quality $> 5$
- **Hypotheses**:
  - $H_0$: There is no difference in the mean alcohol content between Group A and Group B ($\mu_A = \mu_B$).
  - $H_1$: There is a significant difference in the mean alcohol content between Group A and Group B ($\mu_A \neq \mu_B$).

```python
group_a = df_wine[df_wine['quality'] <= 5]['alcohol']
group_b = df_wine[df_wine['quality'] > 5]['alcohol']

print(f"Group A (Quality <= 5) Mean Alcohol: {group_a.mean():.2f}%")
print(f"Group B (Quality > 5) Mean Alcohol: {group_b.mean():.2f}%")

t_stat_w, p_val_w = stats.ttest_ind(group_a, group_b, equal_var=False)
print(f"t-statistic: {t_stat_w:.4f}")
print(f"p-value: {p_val_w:.4e}")
```

# Q3 :

### 1.1 Student Performance Dataset Analysis

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error
```

```python
df_student = pd.read_csv('/content/student-mat.csv', sep=',')
df_student.head()
```

```python
X = df_student[['G1', 'G2', 'studytime']]
y = df_student['G3']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

reg = LinearRegression()
reg.fit(X_train, y_train)

y_pred = reg.predict(X_test)
rmse = root_mean_squared_error(y_test, y_pred)

print("Model Coefficients:")
for col, coef in zip(X.columns, reg.coef_):
    print(f"  {col}: {coef:.4f}")
print(f"Model Intercept: {reg.intercept_:.4f}")
print(f"Test RMSE: {rmse:.4f}")
```

### 3.2 One-Way ANOVA Test (Student Study-Time)
Compare final grades `G3` across the different levels of `studytime` categories.

```python
model_student = ols('G3 ~ C(studytime)', data=df_student).fit()
anova_student = sm.stats.anova_lm(model_student, typ=2)
display(anova_student)
```
