### 1. Write a Python programming to display a bar chart of the popularity of programming Languages.
- Sample data:
- Programming languages: Java, Python, PHP, JavaScript, C#, C++
- Popularity: 22.2, 17.6, 8.8, 8, 7.7, 6.7

```python
Programming_languages = ["Java","Python","PHP","JavaScript","C#","C++"]
Popularity = [22.2, 17.6, 8.8, 8, 7.7, 6.7]
```

```python
plt.bar(Programming_languages, Popularity,color="darkblue", zorder=3,align='edge')
plt.title("Popularity of Programming Language\n Worldwide, Oct 2017 compared to year ago")
plt.xlabel("Languages")
plt.ylabel("Popularity")
plt.xlim(0,6)
plt.ylim(0,25)
plt.minorticks_on()
plt.grid(which="major",linestyle='-',color='r',linewidth= 0.5)
plt.grid(which="minor",linestyle='--',color='black',linewidth= 0.5)
plt.show()
```

### 10. Write a Python program to create bar plot of scores by group and gender. Use multiple X values on the same chart for men and women.
- Sample Data:
- Means (men) = (22, 30, 35, 35, 26)
- Means (women) = (25, 32, 30, 35, 29)


```python
n_groups = 5
men_means = (22, 30, 35, 35, 26)
women_means = (25, 32, 30, 35, 29)
```

```python
men_means
```

```python
women_means
```

```python
index = np.arange(n_groups)
bar_width = 0.35
opacity = 0.8

bar1 = plt.bar(index, men_means, bar_width,alpha=opacity,color='g',label='Men',edgecolor="black")

bar2 = plt.bar(index + bar_width, women_means, bar_width,alpha=opacity,color='r',edgecolor="black",label='Women')

plt.xlabel('Person')
plt.ylabel('Scores')
plt.title('Scores by person')
plt.xticks(index + bar_width / 2, ('G1', 'G2', 'G3', 'G4', 'G5'))
plt.legend()

plt.tight_layout()
plt.show()
```
