```python
from scipy.stats import binom
```

```python
n=5 #number of students asked(trails)
k=3  #number who like python(sucesses)
p=0.66 # probability that one student likes python
```

```python
pmf=binom.pmf(k,n,p)
```

```python
cdf=binom.cdf(k,n,p)
print(f"PMF (P(X={k}))={pmf:.4f}")
print(f"CDF (P(X<={k}))={cdf:.4f}")
```

```python
#PDF
prob = binom.cdf(3,n,p) - binom.cdf(1,n,p)
print(f"Binomial Probability P(2<=X<=3) = {prob:.4f}")
```

```python
# PDF Another Method
pmf_2 = binom.pmf(2,n,p)
pmf_3 = binom.pmf(3,n,p)
prob = pmf_2 + pmf_3
print(f"Binomial Probability P(2<=X<=3) = {prob:.4f}")
```
