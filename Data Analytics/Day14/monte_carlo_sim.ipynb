```python
import numpy as np
```

```python
def roll_dice():
    #simulation of rolling dice twce , min number we get id 1 and max is 6
    #ad  the result tothe two simulation, so possible values will be (1,1),(1,2)....(6,6) sum will between 2 to 12
    # RUn multiple times to verify
    return np.sum(np.random.randint(1,7,2))
```

```python

# Part 2
# 5 dollars if you get 7 and take 1 dollar if you get a number other than 7
#  how do we kn0w what will happen?
# our own monte carllo sim like function

def monte_carlo_simulation(runs=1000):
    result = np.zeros(2)
    for _ in range(runs):
        if roll_dice() == 7:
            result[0] += 1
        else:
            result[1] += 1
    return result
```
