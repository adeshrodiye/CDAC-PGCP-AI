```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pulp import LpInteger, LpMaximize, LpProblem, LpVariable, value
```

```python
model = LpProblem("FactoryProductionProblem",LpMaximize)
```

```python
# Parameters : name, lower limit, upper limi, type 
chairs = LpVariable("Chairs",0,None,LpInteger)
tables = LpVariable("Tables",0,None,LpInteger)
```

```python
model += 300 * (chairs) + 1000 * tables, "Profit"
model += 5 * (chairs) + 10 * tables <= 400, "Wood_Constraint"
model += 2 * chairs + 6 * tables <= 300, "Time_Constraint"
```

```python
model.solve()
for v in model.variables():
    print(v.name, "=", v.varValue)
print("Total Profit : ", value(model.objective))
```

```python

# Car example
model = LpProblem("Car_Production",LpMaximize)
```

```python
x = LpVariable("A_car",0,None,LpInteger)
y = LpVariable("B_car",0,None,LpInteger)
```

```python
model += 30000* x + 45000 * y, "Profit"
```

```python
model += 3*x + 4*y <= 30, "Robot_Time"
model += 5*x + 6*y <= 30, "Engineer_Time"
model += 1.5*x + 3*y <= 21, "Detailer_Time"
```

```python
model.solve()
for v in model.variables():
    print(v.name, "=", v.varValue)
print("Total Profit : ", value(model.objective))
```
