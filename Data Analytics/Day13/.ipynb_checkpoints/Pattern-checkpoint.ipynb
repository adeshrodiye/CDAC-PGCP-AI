```python
import turtle

# Set up the screen
screen = turtle.Screen()
screen.bgcolor("black")  # Set background to black for a neon look

# Create the turtle
t = turtle.Turtle()
t.speed(0)  # Set to maximum drawing speed

# List of vibrant colors
colors = ["red", "purple", "blue", "green", "orange", "yellow"]

# Draw the spiral pattern
for i in range(360):
    t.color(colors[i % len(colors)])  # Cycle through colors
    t.width(i / 100 + 1)             # Gradually increase pen width
    t.forward(i)                     # Move forward by current step size
    t.left(59)                       # Turn left by a precise angle for symmetry

# Keep the window open until clicked
turtle.done()
```
