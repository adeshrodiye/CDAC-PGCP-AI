```python
#HANDLING MISSING VALUES WITH NP.NAN (NOT A NUMBER)
#INDENTIFY AND CLEAN DATA
expenses = np.array([2500,3000,np.nan, 4200,3900,np.nan, 3100])
expenses
```

```python
expenses.sum()
```

```python
missing = np.isnan(expenses)
```

```python
missing
```

```python
clean = expenses[~np.isnan(expenses)]
clean
```

```python
np.nan_to_num(expenses, nan=0)
```

```python
filled = np.where(np.isnan(expenses),np.nanmean(expenses), expenses)
filled.round(2)
```

#### How to handle null
1. DROP THE ROWS
2. REPLACE NULL WITH 0 OR AN IMPOSSIBLE VALUES
3. REPLACE NULL WITH MEAN/MEDIAN
