```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("weight-height.csv")
df
```

```python
cov = df[["Height","Weight"]].cov()
corr = df[["Height","Weight"]].corr()
```

```python
cov
```

```python
corr
```

```python
males = df[df["Gender"]=="Male"]
males
```

```python
Females = df[df["Gender"]=="Female"]
Females
```

```python
male_cov = df[df["Gender"]=="Male"][["Height","Weight"]].cov()
male_cov
```

```python
male_corr = df[df["Gender"]=="Male"][["Height","Weight"]].corr()
male_corr
```

```python
Female_cov = df[df["Gender"]=="Female"][["Height","Weight"]].cov()
Female_cov
```

```python
Female_corr = df[df["Gender"]=="Female"][["Height","Weight"]].corr()
Female_corr
```

```python
plt.figure(figsize=(10,6))
plt.scatter(males['Height'],males['Height'],color='red',label='Height vs Weight')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df['Height'],df['Height'],color='red',alpha=0.5,label='Bitcoin vs Gold')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
