```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('SP500 oil gold bitcoin.csv')
df
```

```python
corr = df[['BITCOIN','Gold','S&P500']].corr()
cov = df[['BITCOIN','Gold','S&P500']].cov()
```

```python
corr
```

```python
cov
```

```python
cov.to_string(float_format='{:.4f}'.format)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df['BITCOIN'],df['Gold'],color='red',alpha=0.5,label='Bitcoin vs Gold')
plt.scatter(df['BITCOIN'],df['S&P500'],color='green',alpha=0.5,label='Bitcoin vs S&P500')
plt.scatter(df['S&P500'],df['Gold'],color='yellow',alpha=0.5,label='S&P500 vs Gold')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Seaborn Equivalent
sns.scatterplot(x="BITCOIN", y= "Gold", data= df, color='red', alpha=0.5, label='Bitcoins vs Gold')
```

```python
sns.heatmap(corr,annot=True, fmt=".2f",cmap="viridis")
plt.title("Overall Correlation Heatmap")
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.show()
```
