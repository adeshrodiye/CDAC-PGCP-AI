## 2. Gradient Boosting on HR_CSV

```python
os.chdir('E:\Backup Files Abhinav_Satyam\Practical Machine Learning\Sanjay Sir Git Repo\Datasets\Cases\human-resources-analytics')
hr = pd.read_csv('HR_comma_sep.csv')
hr
```

```python
X,y = hr.drop('left', axis = 1) , hr['left']

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.3, random_state=26, stratify=y)
```

```python
ohe = OneHotEncoder(sparse_output=False,drop='first').set_output(transform='pandas')
ct = make_column_transformer(
    (ohe, make_column_selector(dtype_include=object)),
    ('passthrough', make_column_selector(dtype_include=['int64','float64'])),
    verbose_feature_names_out=False
).set_output(transform='pandas')
```

```python
X_train_trans = ct.fit_transform(X_train) 
X_test_tans = ct.transform(X_test)
```

```python
gbm = GradientBoostingClassifier(random_state = 26)
gbm.fit(X_train_trans, y_train)
y_pred = gbm.predict(X_test_tans)
accuracy_score(y_test, y_pred)
```

```python
learning_rate = np.linspace(0.01,0.9,10)
max_depth = [None,2,3,4]
n_estimator  = [25,50,100]

scores =  []
for l in tqdm(learning_rate):
    for d in max_depth:
        for n in n_estimator:
            
            gbm = GradientBoostingClassifier(random_state = 26, learning_rate=l, max_depth=d, n_estimators=n)
            gbm.fit(X_train_trans, y_train)
            y_pred = gbm.predict(X_test_tans) 

            scores.append([l, d, n ,accuracy_score(y_test, y_pred) ])
df_scores = pd.DataFrame(scores , columns=['learning_rate','max_depth','n_estimator','score'])
df_scores.sort_values(by='score' , ascending=False)
```
