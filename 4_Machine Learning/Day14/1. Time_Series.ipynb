# Time Series

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../Datasets/monthly-milk-production-pounds-p.csv",index_col=0)
df.head()
```

```python
df.index = pd.to_datetime(df.index).to_period("M")
```

```python
df.index.dtype
```

```python
df
```

```python
df.plot()
plt.title("Monthly Milk Productoion")
plt.xlabel("Months")
plt.ylabel("Production [in pounds]")
plt.show()
```

## resampling

```python
df_q = df.resample("Q").mean() #downsampled
df_q.plot()
plt.title("Quarterly Average Milk Production")
plt.xlabel("Quarters")
plt.ylabel("Production [in pounds]")
plt.show()
```

```python
df_q = df.resample("Y").mean() #downsampled
df_q.plot()
plt.title("Quarterly Average Milk Production")
plt.xlabel("Quarters")
plt.ylabel("Production [in pounds]")
plt.show()
```

```python
df2 = pd.read_csv("../Datasets/Electric_Production.csv",index_col=0)
df2.head()
```

```python
df2.index = pd.to_datetime(df2.index).to_period("M")
df2.index.dtype
```

```python
df2.plot()
plt.title("Monthly Electric Production")
plt.xlabel("Quarters")
plt.ylabel("Production")
plt.show()
```

```python
df_q = df2.resample("Q").mean() #downsampled
df_q.plot()
plt.title("Quarterly Average Milk Production")
plt.xlabel("Quarters")
plt.ylabel("Production [in pounds]")
plt.show()
```

```python
df_y = df2.resample("Y").mean() #downsampled
df_y.plot()
plt.title("Yearly Electricity Production")
plt.xlabel("Year")
plt.ylabel("Production [in pounds]")
plt.show()
```

```python
# Logarithmic graph
df2["IPG2211A2N"] = np.log(df2["IPG2211A2N"]) # change only scale
df2.plot()
plt.title("Log Scaled Electricity Production")
plt.xlabel("Months")
plt.ylabel("Production")
plt.show()
```
