- if series is incr/decr it is non-stationaary
--------
#####  Weak White Noise Model : 
- fixed constant mean, A fixed Var, No corr of any time point value with others [thereotical example of stationary process]
- $Y_t = \epsilon_t$,
-----------
##### Random Walk Model : 
- No specific mean and var, strong dependence over time 
- white noise - changes/incr in RW
- $Y_t = Y_{t-1} + \epsilon_t$
------
##### DIckey Fuller Test

- used to test stationarity of any time series
- adfuller()
- ------


```python
import numpy as np
import pandas as pd
from tqdm import tqdm
import matplotlib.pyplot as plt
import statsmodels.api as sm
from sklearn.metrics import root_mean_squared_error
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.arima.model import ARIMA
```

## FMAC-HPI_24420 Dataset

```python
df = pd.read_csv("../Datasets/FMAC-HPI_24420.csv")
df.head()
```

```python
plot_acf(df["NSA Value"], lags=40, alpha=None,zero=False)
plt.show()
```

```python
res = adfuller(df["NSA Value"], maxlag=10)
print("P VAlue : ", res[1])
if res[1] < 0.05:
    print("Time Series is Stationary")
else:
    print("Time Series is not Stationary")
```

```python
diff_1 = df["NSA Value"].diff().dropna()
res = adfuller(diff_1, maxlag=10)
print("P VAlue : ", res[1])
if res[1] < 0.05:
    print("Time Series is Stationary")
else:
    print("Time Series is not Stationary")
```

```python
diff_2 = diff_1.diff().dropna()
res = adfuller(diff_2, maxlag=10)
print("P VAlue : ", res[1])
if res[1] < 0.05:
    print("Time Series is Stationary")
else:
    print("Time Series is not Stationary")
```
