```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error,r2_score
```

```python
df = pd.read_csv("ChemicalProcess.csv")
df
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
X,y = df.drop("Yield",axis=1), df["Yield"]
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=26, test_size=0.3)
```

```python
imputer = SimpleImputer(strategy="mean").set_output(transform='pandas')
X_trn_imp = imputer.fit_transform(X_train)
X_tst_imp = imputer.transform(X_test)
```

```python
X_trn_imp.isnull().sum()
```

## Linear Regression

```python
lr = LinearRegression()
lr.fit(X_trn_imp,y_train)
y_pred = lr.predict(X_tst_imp)
r2_score(y_test,y_pred)
```

## Ridge

```python
alpha = np.linspace(0.001,10,20)
scores=[]
for i in alpha:
    ridge = Ridge(alpha=i)
    ridge.fit(X_trn_imp,y_train)
    y_pred = ridge.predict(X_tst_imp)
    scores.append([i,r2_score(y_test, y_pred),root_mean_squared_error(y_test,y_pred)])
df_scores = pd.DataFrame(scores, columns=['alpha','r2_score','RMSE'])        
df_scores.sort_values('r2_score')
```

```python
df_scores.nlargest(1,"r2_score")
```

## Lasso

```python
alpha = np.linspace(0.001,10,20)
scores=[]
for i in alpha:
    lasso = Lasso(alpha=i,max_iter=100000)
    lasso.fit(X_trn_imp,y_train)
    y_pred = lasso.predict(X_tst_imp)
    scores.append([i,r2_score(y_test, y_pred),root_mean_squared_error(y_test,y_pred)])
df_scores = pd.DataFrame(scores, columns=['alpha','r2_score','RMSE'])        
df_scores.sort_values('r2_score')
```

```python
df_scores.nlargest(1,"r2_score")
```

## Elastic Net

```python
alpha = np.linspace(0.001,10,20)
ratios = np.linspace(0.001,1,20)
scores = []
for a in alpha:
    for r in ratios:
        el = ElasticNet(alpha=a,l1_ratio=r,max_iter=100000)
        el.fit(X_trn_imp,y_train)
        y_pred = el.predict(X_tst_imp)
        scores.append([a,r,r2_score(y_test, y_pred)])
df_scores = pd.DataFrame(scores, columns=['alpha','l1','r2_score'])        
df_scores.sort_values('r2_score',ascending=False)
```

```python
df_scores.nlargest(1,"r2_score")
```


## Using Pipeline

```python
imputer = SimpleImputer(strategy="mean").set_output(transform='pandas')
lr = LinearRegression()
pipe = Pipeline([("IMP",imputer),("LR",lr)])
pipe.fit(X_train, y_train)
y_pred = pipe.predict(X_test)
r2_score(y_test, y_pred)
```

```python
# Ridge [with pipe]
```

```python
alpha = np.linspace(0.001,10,20)
scores=[]
for i in alpha:
    ridge = Ridge(alpha=i)
    pipe = Pipeline([("IMP",imputer),("Ridge",ridge)])
    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_test)
    scores.append([i,r2_score(y_test, y_pred)])
df_scores = pd.DataFrame(scores, columns=['alpha','r2_score'])        
df_scores.sort_values('r2_score')
```

```python
df_scores.nlargest(1,'r2_score')
```

```python
# Lasso [With Pipe]
```

```python
alpha = np.linspace(0.001,10,20)
scores=[]
for i in alpha:
    lasso = Lasso(alpha=i,max_iter=50000)
    pipe = Pipeline([("IMP",imputer),("lasso",lasso)])
    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_test)
    scores.append([i,r2_score(y_test, y_pred)])
df_scores = pd.DataFrame(scores, columns=['alpha','r2_score'])        
df_scores.sort_values('r2_score')
```

```python
df_scores.nlargest(1,'r2_score')
```
