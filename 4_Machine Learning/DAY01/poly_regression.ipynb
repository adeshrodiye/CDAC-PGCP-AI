```python
# Boston : y=medv, try rmse degree : 2,3,4,5

boston = pd.read_csv("Boston.csv")
boston.head()
```

```python
X = boston.drop("medv",axis=1)
y = boston["medv"]
poly = PolynomialFeatures(degree=2, include_bias=False).set_output(transform="pandas")
X_poly = poly.fit_transform(X)
X_poly
```

```python
X = boston.drop("medv",axis=1)
y = boston["medv"]
poly = PolynomialFeatures(degree=3, include_bias=False).set_output(transform="pandas")
X_poly = poly.fit_transform(X)
X_poly
```

```python
poly = PolynomialFeatures(degree=2, include_bias=False).set_output(transform="pandas")
X,y = boston.drop("medv", axis=1), boston["medv"]
X_poly = poly.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_poly,y,random_state=26,test_size=0.3)

lr = LinearRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
root_mean_squared_error(y_test,y_pred)
```

```python
poly = PolynomialFeatures(degree=3, include_bias=False).set_output(transform="pandas")
X,y = boston.drop("medv", axis=1), boston["medv"]
X_poly = poly.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_poly,y,random_state=26,test_size=0.3)

lr = LinearRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
root_mean_squared_error(y_test,y_pred)
```
