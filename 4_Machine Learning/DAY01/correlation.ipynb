```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error,r2_score
```

```python
pizza = pd.read_csv("../Datasets/pizza.csv")
```

```python
pizza.head()
```

```python
pizza["Promote"].corr(pizza['Sales'])
```

```python
plt.scatter(pizza['Promote'],pizza["Sales"])
plt.xlabel("Promotional Expenses")
plt.ylabel("Sales")
plt.show()
```

```python
import numpy as np

# 1. Calculate the linear fit (slope 'm' and intercept 'b')
m, b = np.polyfit(pizza["Promote"], pizza["Sales"], 1)

# 2. Plot the scatter plot and the trend line
plt.scatter(pizza["Promote"], pizza["Sales"])  #[cite: 1]
plt.plot(
    pizza["Promote"],
    m * pizza["Promote"] + b,
    color="red",
    linewidth=2,
    label="Trend Line",
)

plt.xlabel("Promotional Expenses")  #[cite: 1]
plt.ylabel("Sales")  #[cite: 1]
plt.legend()
plt.show()
```

```python
X,y = boston.drop("medv", axis=1), boston["medv"]
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=26,test_size=0.8)
```

```python
lr = LinearRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
root_mean_squared_error(y_test,y_pred)
```
