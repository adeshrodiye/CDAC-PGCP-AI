```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression, LinearRegression, Ridge, Lasso
from sklearn.ensemble import VotingClassifier, VotingRegressor
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from tqdm import tqdm
```

```python
df = pd.read_csv("Concrete_Data.csv")
df.head()
```

```python
X, y = df.drop("Strength",axis=1), df["Strength"]
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=26,test_size=0.3)
```

```python
dtr = DecisionTreeRegressor(random_state=26)
lr = LinearRegression()
rd = Ridge()
ls = Lasso()
voting = VotingRegressor([("TREE",dtr),("LR",lr),("Ridge",rd),("Lasso",ls)],weights=[1,0.8,0.9,0.9])
voting.fit(X_train,y_train)
```

```python
y_pred = voting.predict(X_test)
r2_score(y_test,y_pred)
```

```python
depths = [None,2,3,4,5]
Ks = [2,5,10]
scores = []
for d in tqdm(depths):
            dtr = DecisionTreeRegressor(random_state=26,max_depth=d)
            lr = LinearRegression()
            rd = Ridge()
            ls = Lasso()
            voting = VotingRegressor([("TREE",dtr),("LR",lr),("Ridge",rd),("Lasso",ls)],weights=[1,0.8,0.9,0.9])
            voting.fit(X_train,y_train)
            y_pred = voting.predict(X_test)
            scores.append([d,r2_score(y_test,y_pred)])
df_scores = pd.DataFrame(scores, columns=["D","Scores"])
df_scores.sort_values("Scores",ascending=False)
```
