```python
import os
os.chdir('E:\Backup Files Abhinav_Satyam\Practical Machine Learning\Sanjay Sir Git Repo\Datasets\Cases\Kyphosis')
import pandas as pd
import numpy as np 

from tqdm import tqdm

import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score, classification_report
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay , roc_curve, RocCurveDisplay, roc_auc_score
```

```python
df = pd.read_csv('Kyphosis.csv')
df
```

```python
X,y = df.drop('Kyphosis', axis = 1) , df['Kyphosis']

le = LabelEncoder()
le_y = le.fit_transform(y)

X_train, X_test, y_train, y_test = train_test_split(X,le_y, test_size = 0.3, random_state=26,stratify=le_y)


```

```python
gbm = GradientBoostingClassifier(random_state = 26)
gbm.fit(X_train, y_train)
y_pred = gbm.predict(X_test)
roc_auc_score(y_test, y_pred)
```

```python
gbm = GradientBoostingClassifier(random_state = 26)
gbm.fit(X_train, y_train)
y_pred_proba = gbm.predict_proba(X_test)
roc_auc_score(y_test, y_pred_proba[:,1])
```

```python

learning_rate = np.linspace(0.01,0.9,10)
max_depth = [None,2,3,4]
n_estimator  = [25,50,100]

scores =  []
for l in tqdm(learning_rate):
    for d in max_depth:
        for n in n_estimator:
            
            gbm = GradientBoostingClassifier(random_state = 26, learning_rate=l, max_depth=d, n_estimators=n)
            gbm.fit(X_train, y_train)
            y_pred_proba = gbm.predict_proba(X_test) 

            scores.append([l, d, n ,roc_auc_score(y_test, y_pred_proba[:,1])])
            
df_scores = pd.DataFrame(scores , columns=['learning_rate','max_depth','n_estimator','roc_auc_score'])
df_scores.sort_values(by='roc_auc_score' , ascending=False)
```
