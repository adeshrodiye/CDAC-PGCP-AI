```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error,r2_score
```

```python
os.chdir("E:/Vishal_Adesh/Machine Learning/Datasets")
```

```python
housing = pd.read_csv("Housing.csv")
housing.head()
```

```python
X, y = housing.drop('price',axis=1),housing['price']
# poly = PolynomialFeatures(include_bias=False)
# X_poly = poly.fit_transform(X)
# X_poly
```

```python
# One hot Encoding categorical features

from sklearn.preprocessing import OneHotEncoder


categorical_features = ["driveway","recroom","fullbase","gashw","airco","prefarea"]

#create instance of onehot encoder
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore',drop="first").set_output(transform='pandas')
num_cols = list(X.select_dtypes(include=['number']).columns)
str_cols = list(X.select_dtypes(include=['object']).columns)
X_num = X[num_cols]
X_str = X[str_cols]
X_ohe = encoder.fit_transform(X_str)
X_enc = pd.concat([X_num,X_ohe],axis=1)


```

```python
str_cols
```

```python
X_enc
```

```python
# encoded_data = encoder.fit_transform(housing[categorical_features])
# encoded_data
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_enc,y,random_state=26,test_size=0.3)
```

```python
ridge = Ridge(alpha=0.5)
ridge.fit(X_train,y_train)
```

```python
ridge.coef_
```

```python
alpha = [0.1,0.25,0.7,1,1.5,2,5,10,50]

for i in alpha:
    ridge = Ridge(alpha=i)
    ridge.fit(X_train,y_train)
    y_pred = ridge.predict(X_test)
    print("\nAlpha : ",i)
    print("ridge.coef : \n",ridge.coef_)
    print("root_mean_squared_error : ",root_mean_squared_error(y_test,y_pred))
```

```python
lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
lr.coef_
```

```python
y_pred = ridge.predict(X_test)
root_mean_squared_error(y_test,y_pred)
```

```python
poly = PolynomialFeatures(degree=2,include_bias=False).set_output(transform='pandas')
X_poly = poly.fit_transform(X_enc)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_poly, y, random_state=26,test_size=0.2)
```

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred = lr.predict(X_test)
root_mean_squared_error(y_test,y_pred)
```

```python
r2_score(y_test,y_pred)
```

```python
poly = PolynomialFeatures(degree=3,include_bias=False).set_output(transform='pandas')
X_poly = poly.fit_transform(X_enc)
X_train, X_test, y_train, y_test = train_test_split(X_poly, y, random_state=26,test_size=0.2)
lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred = lr.predict(X_test)
print(root_mean_squared_error(y_test,y_pred))
print(r2_score(y_test,y_pred))
```

```python
poly = PolynomialFeatures(degree=4,include_bias=False).set_output(transform='pandas')
X_poly = poly.fit_transform(X_enc)
X_train, X_test, y_train, y_test = train_test_split(X_poly, y, random_state=26,test_size=0.2)
lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred = lr.predict(X_test)
print(root_mean_squared_error(y_test,y_pred))
print(r2_score(y_test,y_pred))
```
