```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error,r2_score
```

```python
df = pd.read_csv("ChemicalProcess.csv")
df
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
X,y = df.drop("Yield",axis=1), df["Yield"]
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=26, test_size=0.3)
```

```python
imputer = SimpleImputer(strategy="mean").set_output(transform='pandas')
X_trn_imp = imputer.fit_transform(X_train)
X_tst_imp = imputer.transform(X_test)
```

```python
X_trn_imp.isnull().sum()
```

## Linear Regression

```python
lr = LinearRegression()
lr.fit(X_trn_imp,y_train)
y_pred = lr.predict(X_tst_imp)
r2_score(y_test,y_pred)
```
