```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, r2_score
from sklearn.preprocessing import StandardScaler, MinMaxScaler,OneHotEncoder
from sklearn.compose import make_column_selector,make_column_transformer
from sklearn.preprocessing import OneHotEncoder,LabelEncoder,OrdinalEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.naive_bayes import BernoulliNB
```

```python
df = pd.read_csv("Cancer.csv",index_col=0)
df
```

```python
df.isnull().sum()
```

```python
X, y = df.drop("Class",axis=1), df["Class"]
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=26, test_size=0.3, stratify=y)
```

```python
imputer = SimpleImputer(strategy="most_frequent")
```

```python
ohe = OneHotEncoder(sparse_output=False, drop="first").set_output(transform='pandas')
nb = BernoulliNB()
pipe = Pipeline([("imputer",imputer),("OHE", ohe),("NB",nb)])
pipe.fit(X_train,y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
y_pred
```

```python
oe = OrdinalEncoder().set_output(transform="pandas")
nb = BernoulliNB()
pipe = Pipeline([("imputer",imputer),("OHE", oe),("NB",nb)])
pipe.fit(X_train,y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
encoder_categories = [
    ["20-29", "30-39", "40-49", "50-59", "60-69", "70-79"],  # age
    ["premeno", "ge40", "lt40"],  # menopause
    [
        "0 to 4",
        "5 to 9",
        "10 to 14",
        "15 to 19",
        "20 to 24",
        "25 to 29",
        "30 to 34",
        "35 to 39",
        "40 to 44",
        "45 to 49",
        "50 to 54",
    ],  # tumor-size
    ["0 to 2", "3 to 5", "6 to 8", "9 to 11", "12 to 14", "15 to 17",'24 to 26'],  # inv-nodes
    ["no", "yes",],  # node-caps
    ["one", "two", "three"],  # deg-malig
    ["left", "right"],  # breast
    ["left_up", "left_low", "right_up", "right_low", "central"],  # breast-quad
    ["no", "yes"],  # irradiat
]
```

```python
# Initialize OrdinalEncoder with explicit categories
ordinal_enc = OrdinalEncoder(categories=encoder_categories).set_output(transform="pandas")
nb = BernoulliNB()
pipe = Pipeline([("imputer",imputer),("oe", ordinal_enc),("NB",nb)])
pipe.fit(X_train,y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```
