```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC,SVR
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import accuracy_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.compose import make_column_selector,make_column_transformer
from tqdm import tqdm
```

# SVC for Student Dataset

```python
df = pd.read_csv("student_dataset_10000_rows.csv")
df.head()
```

```python
X, y = df.drop("placement_status",axis=1), df["placement_status"]
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
```

```python
svm = SVC(kernel="linear")
svm.fit(X_train,y_train)
y_pred = svm.predict(X_test)
accuracy_score(y_test,y_pred)
```

# SVC For HR Analytics

```python
hr = pd.read_csv("../Day5/HR_comma_sep.csv")
hr
```

```python
X, y = hr.drop("left",axis=1), hr["left"]
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=26,test_size=0.3,stratify=y)
```

```python
ohc = OneHotEncoder(sparse_output=False, handle_unknown="ignore",drop="first").set_output(transform="pandas")
ct = make_column_transformer((ohc,make_column_selector(dtype_include=object)),
                             ("passthrough", make_column_selector(dtype_include=['int64','float64'])),
                             verbose_feature_names_out=False)
ct = ct.set_output(transform="pandas")
```

```python
scaler = StandardScaler()
svm = SVC(kernel="linear")
pipe = Pipeline([("ct",ct),("scaler",scaler),("svm",svm)])
pipe.fit(X_train, y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
Cs = np.linspace(0.001,10,20)
scores = []
for c in tqdm(Cs):
    svm = SVC(kernel="linear",C=c)
    pipe = Pipeline([("ct",ct),("scaler",scaler),("svm",svm)])
    pipe.fit(X_train,y_train)
    y_pred = pipe.predict(X_test)
    scores.append([c,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(scores, columns=["K","Score"])
df_scores.sort_values("Score",ascending=False)
```

# SVC : Polynomial

```python
scaler = StandardScaler()
svm = SVC(kernel="poly")
pipe = Pipeline([("ct",ct),("scaler",scaler),("svm",svm)])
pipe.fit(X_train, y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
Cs = np.linspace(0.001,10,20)
scores = []
for c in tqdm(Cs):
    for d in [2,3,4]:
        svm = SVC(kernel="poly",C=c,degree=d)
        pipe = Pipeline([("ct",ct),("scaler",scaler),("svm",svm)])
        pipe.fit(X_train,y_train)
        y_pred = pipe.predict(X_test)
        scores.append([c,d,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(scores, columns=["C","D","Score"])
df_scores.sort_values("Score",ascending=False)
```

#  SVM : RBF

```python
scaler = StandardScaler()
svm = SVC(kernel="rbf")
pipe = Pipeline([("ct",ct),("scaler",scaler),("svm",svm)])
pipe.fit(X_train, y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
Cs = np.linspace(0.001,10,20)
Gs = np.linspace(0.1,10,20)
scores = []
for c in tqdm(Cs):
    for g in Gs:
        svm = SVC(kernel="rbf",C=c,gamma=g)
        pipe = Pipeline([("ct",ct),("scaler",scaler),("svm",svm)])
        pipe.fit(X_train,y_train)
        y_pred = pipe.predict(X_test)
        scores.append([c,g,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(scores, columns=["C","gamma","Score"])
df_scores.sort_values("Score",ascending=False)
```
