```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split,KFold, GridSearchCV, RandomizedSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import make_column_transformer, make_column_selector
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.linear_model import ElasticNet
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor
from ISLP import load_data
import warnings
warnings.filterwarnings('ignore', category=FutureWarning)
warnings.filterwarnings('ignore', category=UserWarning)
```

```python
df = load_data('Bikeshare')
df.columns
```

```python
df.head()
```

```python
df.dtypes
```

```python
X = df.drop(columns=['bikers','registered','casual'])
y = df['bikers']
```

```python
df.isnull().sum().sum()
```

```python
ohc = OneHotEncoder(sparse_output=False, handle_unknown="ignore", drop="first")
ct = make_column_transformer((ohc,make_column_selector(dtype_exclude=[int,float])),("passthrough", make_column_selector(dtype_include=[int,float]))).set_output(transform="pandas")
scaler = StandardScaler().set_output(transform="pandas")
```

# ElasticNet

```python
en = ElasticNet(random_state=26)
pipe = Pipeline([("CT",ct),("SCL",scaler),("ElasticNet",en)])
```

```python
pipe.get_params()
```

## GridSearchCV [ElasticNet]

```python
Kfold = KFold(random_state=26,shuffle=True,n_splits=5)
params = { "ElasticNet__alpha" : np.linspace(0.001,10,20),
          "ElasticNet__l1_ratio" : np.linspace(0.001,1,20)}
gcv = GridSearchCV(pipe, param_grid=params, cv=Kfold, scoring="r2", n_jobs=1)
gcv.fit(X, y)
print("Best Params: ", gcv.best_params_)
print("Best Score: ", gcv.best_score_)
```

## RandomizedSearchCV [ElasticNet]

```python
Kfold = KFold(random_state=26,shuffle=True,n_splits=5)
params = { "ElasticNet__alpha" : np.linspace(0.001,10,20),
          "ElasticNet__l1_ratio" : np.linspace(0.001,1,20)}
rcv = RandomizedSearchCV(pipe, param_distributions=params,n_iter=25, cv=Kfold, scoring="r2", n_jobs=1)
rcv.fit(X, y)
print("Best Params: ", rcv.best_params_)
print("Best Score: ", rcv.best_score_)
```

----------

# Random Forest

```python
rf = RandomForestRegressor()
pipe = Pipeline([("CT",ct),("SCL",scaler),("RF",rf)])
```

```python
pipe.get_params()
```

## GridSearchCV [RandomForestRegressor]

```python
params = {"RF__n_estimators": [50, 100, 200, 300],
             "RF__max_depth": [None, 10, 20, 30],
             "RF__min_samples_split": [2, 5, 10],
             "RF__min_samples_leaf": [1, 2, 4]}

gcv = GridSearchCV(pipe, param_grid=params, cv=Kfold, scoring="r2", n_jobs=1)
gcv.fit(X, y)
print("Best Params: ", gcv.best_params_)
print("Best Score: ", gcv.best_score_)
```

## RandomizedSearchCV [RandomForestRegressor]

```python
rcv = RandomizedSearchCV(pipe, param_distributions=params,n_iter=25, cv=Kfold, scoring="r2", n_jobs=1)
rcv.fit(X, y)
print("Best Params: ", rcv.best_params_)
print("Best Score: ", rcv.best_score_)
```

# GradientBoostingRegressor

```python
gbm = GradientBoostingRegressor()
pipe = Pipeline([("CT",ct),("SCL",scaler),("GBM",gbm)])
```
