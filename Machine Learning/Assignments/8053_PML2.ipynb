```python
import numpy as np
import pandas as pd
import os
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.compose import make_column_selector,make_column_transformer
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis,QuadraticDiscriminantAnalysis
from sklearn.svm import SVC
```

```python
df = pd.read_csv('drug200.csv')
df.head()
```

## Logistic Regression

```python
X,y = df.drop('Drug',axis=1),df['Drug']
```

```python
ohe = OneHotEncoder(sparse_output=False,drop='first').set_output(transform='pandas')
ct = make_column_transformer((ohe,make_column_selector(dtype_include='object')),
                            ('passthrough',make_column_selector(dtype_include=['int64','float64'])),
                            verbose_feature_names_out=False)
ct = ct.set_output(transform='pandas')
```

```python
X_train,X_test,y_train,y_test  = train_test_split(X,y,test_size=0.3,random_state=26,stratify=y)
lr  = LogisticRegression(max_iter=2000)
pipe = Pipeline([('OHE',ct),("LR",lr)])
pipe.fit(X_train,y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
solvers=['lbfgs', 'liblinear', 'newton-cg', 'newton-cholesky', 'sag', 'saga']
scores = []
for s in solvers:
    lr  = LogisticRegression(max_iter=2000,solver=s)
    pipe = Pipeline([('OHE',ct),("LR",lr)])
    pipe.fit(X_train,y_train)
    y_pred = pipe.predict(X_test)
    scores.append([s,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(scores,columns=['solver','accuracy'])
df_scores.sort_values('accuracy',ascending=False)
```

## KNN

```python
knn = KNeighborsClassifier()
scaler = StandardScaler()
pipe = Pipeline([('OHE',ct),('SCL',scaler),("KNN",knn)])
pipe.fit(X_train,y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
neighbors = np.arange(1,16)
scores = []
for n in neighbors:
    knn = KNeighborsClassifier(n_neighbors=n)
    pipe = Pipeline([('OHE',ct),('SCL',scaler),("KNN",knn)])
    pipe.fit(X_train,y_train)
    y_pred = pipe.predict(X_test)
    scores.append([n,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(scores,columns=['n_neighbors','accuracy'])
df_scores.sort_values('accuracy',ascending=False)
```

## Linear Discriminant

```python
lda = LinearDiscriminantAnalysis()
pipe = Pipeline([('OHE',ct),('LDA',lda)])
pipe.fit(X_train,y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
solvers = ['svd', 'lsqr', 'eigen']
scores = []
for s in solvers:
    lda = LinearDiscriminantAnalysis(solver=s)
    pipe = Pipeline([('OHE',ct),('LDA',lda)])
    pipe.fit(X_train,y_train)
    y_pred = pipe.predict(X_test)
    scores.append([s,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(scores,columns=['solver','accuracy'])
df_scores.sort_values('accuracy',ascending=False)
```

## QDA

```python
lda = QuadraticDiscriminantAnalysis()
pipe = Pipeline([('OHE',ct),('LDA',lda)])
pipe.fit(X_train,y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
params = [0.1,0.3,0.5,0.7,0.9]
scores = []
for r in params:
    lda = QuadraticDiscriminantAnalysis(reg_param=r)
    pipe = Pipeline([('OHE',ct),('LDA',lda)])
    pipe.fit(X_train,y_train)
    y_pred = pipe.predict(X_test)
    scores.append([r,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(scores,columns=['reg param','accuracy'])
df_scores.sort_values('accuracy',ascending=False)
```

## SVM

```python
svm = SVC()
scaler = StandardScaler()
pipe = Pipeline([('OHE',ct),('SCL',scaler),('SVM',svm)])
pipe.fit(X_train,y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
kernels = ['linear', 'poly', 'rbf', 'sigmoid']
scores = []
for k in kernels:
    svm = SVC(kernel=k)
    pipe = Pipeline([('OHE',ct),('SCL',scaler),('SVM',svm)])
    pipe.fit(X_train,y_train)
    y_pred = pipe.predict(X_test)
    scores.append([k,accuracy_score(y_test,y_pred)])
df_scores = pd.DataFrame(scores,columns=['kernel','accuracy'])
df_scores.sort_values('accuracy',ascending=False)
```

### The model eith the best accuracy is QuadraticDiscriminantAnalysis(reg_param=0.1)
