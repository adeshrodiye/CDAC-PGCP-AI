Linear Regression | Ridge Regression | Lasso Regression | Elastic net Regression









```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from sklearn.linear_model import LinearRegression, Ridge, ElasticNet, Lasso
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error,r2_score
```

```python
df = pd.read_csv("E:/Vishal_Adesh/Machine Learning/Datasets/Exp_Salaries.csv")
df.head()
```

```python
X = df.drop("Salary",axis=1)
y = df["Salary"]
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore',drop="first").set_output(transform='pandas')
num_cols = list(X.select_dtypes(include=['number']).columns)
str_cols = list(X.select_dtypes(include=['object']).columns)
X_num = X[num_cols]
X_str = X[str_cols]
X_ohe = encoder.fit_transform(X_str)
X_enc = pd.concat([X_num,X_ohe],axis=1)
```

```python
X_enc.head()
```

# Linear Regression

```python
X_train, X_test, y_train, y_test = train_test_split(X_enc,y,random_state=26,test_size=0.3)
```

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred = lr.predict(X_test)
```

```python
r2 = r2_score(y_test, y_pred)
r2
```

```python
root_mean_squared_error(y_test,y_pred)
```

```python
lr.coef_
```

# Ridge

```python
alpha = np.linspace(0.001,10,20)
scores=[]
for i in alpha:
    ridge = Ridge(alpha=i)
    ridge.fit(X_train,y_train)
    y_pred = ridge.predict(X_test)
    scores.append([i,r2_score(y_test, y_pred),root_mean_squared_error(y_test,y_pred)])
df_scores = pd.DataFrame(scores, columns=['alpha','r2_score','RMSE'])        
df_scores.sort_values('r2_score')
```

```python
df_scores.nlargest(1,'r2_score')
```

#  Lasso

```python
alpha = np.linspace(0.001,10,20)
scores=[]
for i in alpha:
    lasso = Lasso(alpha=i)
    lasso.fit(X_train,y_train)
    y_pred = lasso.predict(X_test)
    scores.append([i,r2_score(y_test, y_pred),root_mean_squared_error(y_test,y_pred)])
df_scores = pd.DataFrame(scores, columns=['alpha','r2_score','RMSE'])        
df_scores.sort_values('r2_score')
```

```python
df_scores.nlargest(1,'r2_score')
```

# E-Net

```python
alpha = np.linspace(0.001,10,20)
ratios = np.linspace(0.001,1,20)
scores = []
for a in alpha:
    for r in ratios:
        el = ElasticNet(alpha=a,l1_ratio=r)
        el.fit(X_train,y_train)
        y_pred = el.predict(X_test)
        scores.append([a,r,r2_score(y_test, y_pred)])
df_scores = pd.DataFrame(scores, columns=['alpha','l1','r2_score'])        
df_scores.sort_values('r2_score',ascending=False)
```

```python
df_scores.nlargest(1,'r2_score')
```
