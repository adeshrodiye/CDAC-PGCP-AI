# Decorators
- help to modify behavior of f() without actually modifying f() [without changing their actual code ]
- take f() as arg and ret new f() (wrapper)
- Application : Logging, authentication and memoization

```python
import time

def keep_time(inner):
    def wrapper():
        start = time.time()
        inner()
        end = time.time()
        print(f'Time taken : {end-start}')
    return wrapper

@keep_time
def print_square():
    for i in range(100):
        print(i*i)
print_square()
```

```python

def smart_math(inner):
    def wrapper(*args,**kwargs):
        a,b = args
        if a<b:
            a,b = b,a
        return(inner(a,b))
    return wrapper

@smart_math
def subtract(n1,n2):
    return n1-n2

print(subtract(6,89))
```
