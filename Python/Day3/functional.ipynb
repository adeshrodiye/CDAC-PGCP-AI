```python
#lambda argument(s) : expression : lambda x: x * 3
#filter(function, iterable) : filter(lambda x: x % 2 == 0, list1)
```

```python
books = [["Python for Beginers",800],
         ["Head first Java", 1000],
         ["Data Science using Pyhton", 1200]]
books
```

```python
change_case = list(map(lambda book:[book[0].title(), book[1]], books)) #title capitalize first letter of each word
change_case
```

```python
greater_than = list(filter(lambda book: book[1]>900,books))
greater_than
```

```python
sort_by_price = sorted(books, key=lambda book:book[1])
sort_by_price
```

```python
min_price = min(books, key = lambda book:book[1])
min_price
```

```python
colors_data = [{'Name':'Yellow','Rating':50},
               {'Name':'Pink','Rating':70},
               {'Name':'Blue','Rating':60},
               {'Name':'Red','Rating':80},
               {'Name':'Black','Rating':85}]
               
```

```python
least_fav = min(colors_data, key = lambda color: color['Rating'])
least_fav
```

```python
sort_by_rating = sorted(colors_data, key=lambda color:color['Rating']) #Dictionaries use named string keys to access value
sort_by_rating 
```

```python
filtered = list(filter(lambda color:color['Rating']>60, colors_data))
filtered
```
