```python

def add(a,b):
    print(f'Addition: {a+b}')

def subtract(a,b):
    print(f'Subtraction : {a-b}')

def multiply(a,b):
    print(f'Multiplication : {a*b}')

def divide(a,b):
    print(f'Division : {a/b}')
    
while True:
    
    operation = int(input("\n1.Addition\n2.Subtraction\n3.Multiplication\n4.Division\n5.End\n"))
    print("-----------------------")
    if operation != 5:
        a = int(input("Enter 1st number:"))
        b = int(input("Enter 2nd number:"))
    
    match operation:
        case 1:
            add(a,b)
        case 2:
            subtract(a,b)
        case 3:
            multiply(a,b)
        case 4:
            divide(a,b)
        case 5:
            break
    
    



```
