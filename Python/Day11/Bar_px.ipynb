```python
import plotly.express as px
```

```python
dir(px.data)
```

```python
df = px.data.gapminder()
df
```

```python
filt = df['continent']=='Asia'
df_f = df.loc[filt]
df_f
```

```python
fig = px.bar(df_f,x='country',y='lifeExp',color='country',
             animation_frame = 'year')
fig.update_layout(title="Life Exp Over The Years",
                  xaxis_title="Country",
                  yaxis_title="Life Expectancy",)
fig
```

```python
fig = px.bar(df_f,x='country',y='lifeExp',color='country',
             animation_frame = 'year',text_auto='.2s')
fig.update_layout(title="Life Exp Over The Years",
                  xaxis_title="Country",
                  yaxis_title="Life Expectancy",)
fig
```
