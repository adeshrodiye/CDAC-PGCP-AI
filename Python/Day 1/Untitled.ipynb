```python
#indexing and slicing

print(s[4])
print(s[-7])
print(s[5:15])
print(s[:10])
print(s[10:])
print(s[:10]+ s[10:])
print(s[15:5])
print(s[:15:5])
print(s[-15:-2])
print(s[-15:-2:-1])
print(s[-15:-2:2])
print(s[1:15:-1])
print(s[::-1])
```
