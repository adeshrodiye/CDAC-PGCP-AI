```python
num = int(input("Enter a value : "))
if num>100:
    print("Greater than 100")
elif num == 100:
    print("Number is 100")
else:
    print("Less than 100")
print("End")
```

```python
age = int(input("Enter your age : "))
gender = input("Enter your gender M/F : ")

if age> 18:
    if gender == 'M':
        print("Man")
    else:
        print("Women")
else:
    print("Child")
```

```python
marks = int(input("Enter Marks : "))

if 85 < marks <= 100:
    print("A")
if 60 < marks <= 85:
    print("B")
if 45 < marks <= 60:
    print("C")
else:
    print("F")
```
