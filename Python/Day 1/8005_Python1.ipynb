### 1. Using for loop, write and run a Python program for this algorithm. Here is an algorithm to print out n! from 0! to 10! 

```python
factorial = 1

for n in range(11):
    if n == 0:
        factorial = 1
    else:
        factorial *= n
    print(f"{n}! = {factorial}")
```

```python
def isPrime(num):
        if num == 1:
            return 0            
        else:
            for i in range(2,num):
                if num % i == 0:
                    break
            else:
                 print(num)

for num in range(0,11):
    isPrime(num)
```

### 4. Count Digits, Even/Odd, Sum
- e.g. 23456
- output digits : 5
- sum : 20
- Even digits : 3
- odd digits :2

```python
#count
num = 345
count = 0
total_sum = 0
even_count = 0
odd_count = 0

temp = abs(num)
if temp == 0:
    count = 1
else: 
    while temp > 0:
        digit = temp % 10  
        count += 1
        total_sum += digit

        if digit % 2 == 0:
            even_count += 1
        else:
            odd_count += 1

        temp = temp // 10
        
print(f"output digits : {count}")
print(f"sum : {total_sum}")
print(f"Even digits : {even_count}")
print(f"odd digits : {odd_count}")

```

### 5. Find LCM and GCD for given numbers [take 2 numbers] using loops only

```python
a = 12
b = 18

min_num = min(a, b)
gcd = 1
for i in range(min_num, 0, -1):
    if a % i == 0 and b % i == 0:
        gcd = i
        break

max_num = max(a, b)
lcm = max_num
while True:
    if lcm % a == 0 and lcm % b == 0:
        break
    lcm += max_num  

print(f"GCD: {gcd}")  
print(f"LCM: {lcm}")  
```
