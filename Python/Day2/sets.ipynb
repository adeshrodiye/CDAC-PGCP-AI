# Sets
- set is an unordered collection.
- .pop() method on a set removes and returns an arbitrary element
- 

```python
colors = {'red','pink','white','black','blue','white'}
```

```python
for color in colors:
    print(color.upper())
```

```python
print('red' in colors)
```

```python
print('white' not in colors)
```

```python
colors.add('yellow')
colors
```

```python
colors.pop()
colors
```

```python
colors.remove('white')
colors
```

```python
colors.add('white')
```

```python
colors
```

```python
colors.discard('white')
colors
```

```python
colors1 = {'red','yellow','pink','purple','black','orange'}
colors1
```

```python
u_set = colors.union(colors1)
print(u_set)
```

```python
u_set = colors | colors1
print(u_set)
```

```python
i_set = colors.intersection(colors1) 
print(i_set)
```

```python
i_set = colors & colors1
i_set
```

```python
d_set = colors.difference(colors1) #Returns all elements that are in set colors but not in set colors1
d_set
```

```python
d_set = colors1.difference(colors) #Returns all elements that are in set colors1 but not in set colors
d_set
```

```python
d_set = colors - colors1
d_set
```

```python
s_set = colors.symmetric_difference(colors1) #Returns all elements that are in either colors or colors1
s_set
```

```python
s_set = colors ^ colors1
s_set
```
