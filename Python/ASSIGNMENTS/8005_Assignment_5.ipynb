### 1. Draw line chart to show urban and rural population of India over the years using population.csv [rural 	population is the difference between total population and urban population]

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("Population.csv")
df
```

```python
df.isna()
```

```python
df["Rural Population"] = df["Population"]- df["Urban Population"]
df
```

```python
df.drop(index=[14,15,16],inplace=True)
```

```python
df
```

```python
plt.plot(df["Year"],df["Urban Population"],color = "Blue", ls="--",lw=0.8, label="Urban Population")
plt.plot(df["Year"],df["Rural Population"],color = "Red" , ls="-", lw=0.8, label="Rural Population")
plt.xlabel("Year")
plt.ylabel("Population in Billion")
plt.title("Population over time [1960-2025]")
plt.legend()
plt.show()
```

### 2. Using a Bar chart display India's total, urban and rural population from year 2020 to 2025

```python
df = pd.read_csv("Population.csv")
df
```

```python
df["Rural Population"] = df["Population"]- df["Urban Population"]
df
```

```python
df1 = df[df["Year"]>=2020].reset_index(drop=True)
df1
```

```python
x_vals = np.arange(len(df1))
b_width = 0.25
x_vals
```

```python
plt.figure(figsize=(12,8), dpi=100)

plt.bar(x_vals-b_width, df1["Population"], width = b_width, color = "#5b34eb", label = "Total Population")
plt.bar(x_vals, df1["Urban Population"], width = b_width, color = "#3f85e0", label = "Urban Population")
plt.bar(x_vals+b_width, df1["Rural Population"], width = b_width, color = "Pink", label = "Rural Population")

plt.xticks(x_vals,df1["Year"])
plt.xlabel("Years")
plt.ylabel("Population")
plt.title("Population Distribution")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout() 
plt.show()
```

### 3. Using gapminder2007.csv show life expectancy of top 10 highly populated countries from Asia

```python
df = pd.read_csv("gapminder2007.csv")
```

```python
df
```

```python
top = df.loc[df["continent"]=="Asia"].nlargest(10,columns="pop")
top
```

```python
top = top.sort_values("lifeExp",ascending=False)
```

```python
top
```

```python
plt.figure(figsize=(10,6), dpi=100)
bars = plt.bar(top["country"],top["lifeExp"],color="#7d48f0",width = 0.8, label = "Country")
plt.bar_label(bars, padding=3)
plt.xticks(rotation=90)
plt.xlabel("Country")
plt.ylabel("Life Expectency")
plt.title("life expectancy of top 10 highly populated countries from Asia")
plt.legend()
plt.show()
```

### 4. Using a pie chart show distribution of top six selling cars.

```python
df = pd.read_csv("Car_sales.csv") #used kaggle dataset
df
```

```python
df.columns
```

```python
top = df.nlargest(6,"Sales_in_thousands")
top
```

```python
top["Name"] = top["Manufacturer"]+" "+top["Model"]
top
```

```python
plt.figure(figsize=(10,8))
plt.pie(top["Sales_in_thousands"],
        explode=[0.1,0,0,0,0,0],
       labels = top["Name"],
       autopct = "%1.1f%%",
       shadow=True)
plt.title("Top 6 Cars Sales Distribution")
        
```
