```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
df = pd.read_csv("tips.csv")
df.head()
```

```python
total_bill_series = df["total_bill"]
total_bill_series
```

```python
#Normality test with shapiro-wilk test
shapiro = stats.shapiro(total_bill_series)
print(f"Test Statistics (W) : {shapiro.statistic}")
print(f"P-Value : {shapiro.pvalue}")
```

```python
alpha = 0.05
if shapiro.pvalue > alpha:
    print("Data is normally distributed")
else:
    print("Data is not normally distributed")
```

```python
# QQ Plot

stats.probplot(total_bill_series, dist="norm",plot = plt)
plt.title("QQ Plot of Total Bill")
plt.xlabel("Theoretical Quantiles (Normal Distribution)")
plt.ylabel("Ordered Values (Total Bill)")
plt.grid(True)
plt.show()
```
