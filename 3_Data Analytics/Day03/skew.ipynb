```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Day3/airline_passenger_satisfaction.csv")
```

```python
df
```

```python
df.isnull().sum()
```

```python
on_board_services = df["On-board Service"]
```

```python
mean_rating = on_board_services.mean()
```

```python
median_rating = on_board_services.median()
```

```python
skewness_rating = on_board_services.skew()
```

```python
kurtosis_rating = on_board_services.kurt()
```

```python
print(on_board_services.value_counts().sort_index())
print(f"Mean on board Service : {mean_rating}")
```

```python
print(f"Median on board Service : {median_rating}")
print(f"skewness_rating on board Service : {skewness_rating}")
```

```python
print(f"kurtosis_rating on board Service : {kurtosis_rating}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.countplot(x=on_board_services, color="skyblue")
plt.axvline(mean_rating, color='r', linestyle='--',label= f" Median :{median_rating:.2f}")
```
