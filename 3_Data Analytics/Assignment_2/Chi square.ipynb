```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency
```

```python
df = pd.read_csv("Amazon Sale Report.csv")
df.head()
```

```python
df.columns
```

```python
cancelled_pending_orders = df[df["Status"].isin(["Cancelled","Pending"])]
cancelled_pending_orders
```

```python
#Create a contingency table
contigency_table= pd.crosstab(cancelled_pending_orders['Status'],cancelled_pending_orders['Fulfilment'])
contigency_table
```

```python
#Perfrom chi-square test
chi1 , p , dof , ex = chi2_contingency(contigency_table)
```

```python
# Perfrom chi-square test
chi1,p,dof,ex=chi2_contingency(contigency_table)
```

```python
print("Chi-Squared statistic:",chi1)
```

```python
print("p-value",p)
```

```python
print("Degrees of freedom:",dof)
print('Expected valuse:',ex)
if p<0.05:
    print("Reject null hypothesis.")
else:
    print("Fail to reject null hypothesis")
```
