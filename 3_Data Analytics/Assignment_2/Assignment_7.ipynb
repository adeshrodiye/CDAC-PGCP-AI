```python
#4
df = pd.read_csv('fdata.csv')
plt.plot(df['Date'],df['Open'],label = 'Open')
plt.plot(df['Date'],df['High'],label = 'High')
plt.plot(df['Date'],df['Low'],label = 'Low')
plt.plot(df['Date'],df['Close'],label = 'Close')
plt.legend()
```

```python
#5
x = [10, 20, 30]
y1 = [20, 40, 10]
y2 = [40, 10, 30]

plt.plot(x,y1,color='purple',label='line1')
plt.plot(x,y2,color='green',label='line2')
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.xlim(10,30)
plt.ylim(10,40)
plt.legend()
plt.title('Two or more lines on same plot with suitable legends')
```

```python
#6
x=[10,20,30]
y1=[20,40,10]
y2=[40,10,30]
plt.plot(x,y1,color='blue',label='line1-width-3',linewidth=3)
plt.plot(x,y2,color='red',label='line2-width-3',linewidth=5)
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.xlim(10,30)
plt.ylim(10,40)
plt.legend()
plt.title('Two or more lines with different widths and colors with suitable legend')
```

```python
#7
x=[10,20,30]
y1=[20,40,10]
y2=[40,10,30]
plt.plot(x,y1,color='blue',label='line1-dotted',linewidth=3,linestyle='dotted')
plt.plot(x,y2,color='red',label='line2-dashed',linewidth=5,linestyle='dashed')
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.xlim(10,30)
plt.ylim(10,40)
plt.legend()
plt.title('Plot with two or more lines with different styles')
```

```python
#8
x=[1,4,5,6,7]
y=[2,6,3,6,3]
plt.plot(x,y,color='red',linestyle=':',marker='o',markerfacecolor='blue',markeredgecolor="blue",markersize=8)
plt.xlim(1,8)
plt.ylim(1,8)
plt.xlabel('x-axis')
plt.ylabel('y-axis')

plt.title('Display marker')

```

```python
#11
t = np.arange(0.0, 5.0, 0.2)
plt.plot(t, t, "--",color = 'green')
plt.plot(t, t**2, "s" ,color = 'blue')
plt.plot(t, t**3, "^",color='red')
plt.ylim(0,120)
plt.xlim(0,5)
plt.show()
```

```python
#12
dates = [
    "2016-10-03",
    "2016-10-04",
    "2016-10-05",
    "2016-10-06",
    "2016-10-07",
]
close_prices = [772.559998, 776.429993, 776.469971, 776.859985, 775.080017]

df = pd.DataFrame({"Date": dates, "Close": close_prices})

df["Date"] = pd.to_datetime(df["Date"])

plt.plot(
    df["Date"],
    df["Close"],
    color="red",
    linestyle="-",
    linewidth=0.8,
    marker="o",
    markerfacecolor="darkred",
    markeredgecolor="darkred",
    markersize=4,
)

plt.xlim(df["Date"].min(), df["Date"].max())
plt.xticks(["2016-10-03","2016-10-04", "2016-10-05", "2016-10-06" ,"2016-10-07"])
plt.ylim(772.5, 777.0)

plt.show()
```

```python
#13

plt.plot(
    df["Date"],
    df["Close"],
    color="red",
    linewidth=0.8,
    marker="o",
    markerfacecolor="red",
    markersize=3,
)

plt.xlabel("Date")
plt.ylabel("Closing Value")
plt.title("Closing stock value of Alphabet Inc.")
plt.xlim(df["Date"].min(), df["Date"].max())
plt.ylim(772.5, 777.0)
plt.xticks(["2016-10-03","2016-10-04", "2016-10-05", "2016-10-06" ,"2016-10-07"])


plt.grid(True, linestyle="-", linewidth=0.5, color="blue")

plt.show()
```

```python
#14
plt.plot(
    df["Date"],
    df["Close"],
    color="red",
    linewidth=1,
    marker="o",
    markerfacecolor="red",
    markeredgecolor = 'blue',
    markersize=4,
)

plt.xlabel("Date")
plt.ylabel("Closing Value")
plt.title("Closing stock value of Alphabet Inc.")

plt.xlim(df["Date"].min(), df["Date"].max())
plt.ylim(772.5, 777.0)
plt.minorticks_on()

plt.grid(which="major", linestyle="-", linewidth=0.5, color="red")

plt.grid(which="minor", linestyle=":", linewidth=0.5, color="gray")
plt.xticks(["2016-10-03","2016-10-04", "2016-10-05", "2016-10-06" ,"2016-10-07"])


plt.show()
```
