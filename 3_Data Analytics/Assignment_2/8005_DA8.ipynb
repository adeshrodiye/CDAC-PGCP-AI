### 20. Write a Pandas program to subtract two timestamps of same time zone or different time zone.

### 27. Write a Pandas program to calculate one, two, three business day(s) from a specified date. Also find the next business month end from a specific date.

```python
import pandas as pd

date = pd.Timestamp("2023-10-13")

bday_1 = date + pd.offsets.BusinessDay(1)
bday_2 = date + pd.offsets.BusinessDay(2)
bday_3 = date + pd.offsets.BusinessDay(3)
next_bmes = date + pd.offsets.BMonthEnd()

print("Original Date:", date.date())
print("1 Business Day Later:", bday_1.date())
print("2 Business Days Later:", bday_2.date())
print("3 Business Days Later:", bday_3.date())
print("Next Business Month End:", next_bmes.date())
```
