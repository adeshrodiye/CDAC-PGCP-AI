### 14. Write a Python program to create a stacked bar plot with error bars. Note: Use bottom to stack the women?s bars on top of the men?s bars.

- Means (women) = (25, 32, 30, 35, 29)

### 15. Write a Python program to create a horizontal bar chart with differently ordered colors. Note: Use bottom to stack the women?s bars on top of the men?s bars.
- Sample Data Set:
- languages = [
- ['Language','Science','Math'],
- ['Science','Math','Language'],
- ['Math','Language','Science']]
  
- numbers = [
- {'Language':75, 'Science':88, 'Math':96},
- {'Language':71, 'Science':95, 'Math':92},
- {'Language':75, 'Science':90, 'Math':89}]

```python
plt.figure(figsize=(15,4))
languages = [['Language', 'Science', 'Math'],
             ['Science', 'Math', 'Language'],
             ['Math', 'Language', 'Science']]

numbers = [{'Language': 75, 'Science': 88, 'Math': 96},
           {'Language': 71, 'Science': 95, 'Math': 92},
           {'Language': 75, 'Science': 90, 'Math': 89}]


colors = {'Language': 'r', 'Science': 'b', 'Math': 'g'}

students = ['Student-1', 'Student-2', 'Student-3']

y_pos = [0, 1, 2]

fig, ax = plt.subplots()

for i, (student_langs, student_scores) in enumerate(zip(languages, numbers)):
    left_offset = 0
    for lang in student_langs:
        val = student_scores[lang]
        ax.barh(i, val, left=left_offset, color=colors[lang], edgecolor='black', height=0.7)
        left_offset += val

# Set labels and ticks
ax.set_yticks(y_pos)
ax.set_yticklabels(students)
plt.xlim(0, 300)

# Create custom legend entries
handles = [plt.Rectangle((0, 0), 1, 1, color=colors[l], ec="black") for l in ['Language', 'Math', 'Science']]
legend = ax.legend(handles, ['Language', 'Math', 'Science'], loc='upper left', bbox_to_anchor=(1.02, 1.0), borderaxespad=0)

# Grid settings
plt.minorticks_on()
plt.grid(which='major', linestyle='-', linewidth=0.5, color='grey', alpha=0.5)
plt.grid(which='minor', linestyle=':', linewidth=0.5, color='grey', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
languages = [['Language', 'Science', 'Math'],['Science', 'Math', 'Language'],['Math', 'Language', 'Science']]

numbers = [{'Language': 75, 'Science': 88, 'Math': 96},{'Language': 71, 'Science': 95, 'Math': 92},{'Language': 75, 'Science': 90, 'Math': 89}]

colors = {'Language': 'r', 'Science': 'b', 'Math': 'g'}
students = ['Student-1', 'Student-2', 'Student-3']
unique_langs = ['Language', 'Math', 'Science']

flat_langs = np.array(languages).flatten()
student_indices = np.repeat(np.arange(3), 3)
vals = np.array([num[lang] for num in numbers for lang in ['Language', 'Science', 'Math']]).reshape(3, 3)
flat_vals = np.take_along_axis(
    np.array([[num[lang] for lang in ['Language', 'Science', 'Math']] for num in numbers]),
    np.array([[ ['Language', 'Science', 'Math'].index(l) for l in langs ] for langs in languages]),
    axis=1
).flatten()

left_offsets = np.concatenate(([0], np.cumsum(flat_vals)[:-1])) - np.repeat(np.concatenate(([0], np.cumsum(flat_vals)[2::3][:-1])), 3)
segment_colors = np.vectorize(colors.get)(flat_langs)
plt.figure(figsize=(5,5))
fig, ax = plt.subplots(figsize=(10, 4))
ax.barh(student_indices, flat_vals, left=left_offsets, color=segment_colors, edgecolor='black', height=0.7)
ax.set_yticks(np.arange(3))
ax.set_yticklabels(students)
ax.set_xlim(0, 300)
handles = list(map(lambda l: plt.Rectangle((0, 0), 1, 1, color=colors[l], ec="black"), unique_langs))
ax.legend(handles, unique_langs, loc='upper left', bbox_to_anchor=(1.02, 1.0), borderaxespad=0)
plt.minorticks_on()
plt.grid(which='major', linestyle='-', linewidth=0.5, color='grey', alpha=0.5)
plt.grid(which='minor', linestyle=':', linewidth=0.5, color='grey', alpha=0.5)
plt.tight_layout()
plt.show()
```

### 16. Write a Python program to create stack bar plot and add label to each section.
- Sample data:
- people = ('G1','G2','G3','G4','G5','G6','G7','G8')
- segments = 4
- data =
- [[ 3.40022085, 7.70632498, 6.4097905, 10.51648577, 7.5330039,7.1123587, 12.77792868, 3.44773477],
- [ 11.24811149, 5.03778215, 6.65808464, 12.32220677, 7.45964195,6.79685302, 7.24578743, 3.69371847],
- [ 3.94253354, 4.74763549, 11.73529246, 4.6465543, 12.9952182,4.63832778, 11.16849999, 8.56883433],
- [ 4.24409799, 12.71746612, 11.3772169, 9.00514257, 10.47084185,10.97567589, 3.98287652, 8.80552122]]

```python
people = ('G1', 'G2', 'G3', 'G4', 'G5', 'G6', 'G7', 'G8')

data = np.array([
    [3.40022085, 7.70632498, 6.4097905, 10.51648577, 7.5330039, 7.1123587, 12.77792868, 3.44773477],
    [11.24811149, 5.03778215, 6.65808464, 12.32220677, 7.45964195, 6.79685302, 7.24578743, 3.69371847],
    [3.94253354, 4.74763549, 11.73529246, 4.6465543, 12.9952182, 4.63832778, 11.16849999, 8.56883433],
    [4.24409799, 12.71746612, 11.3772169, 9.00514257, 10.47084185, 10.97567589, 3.98287652, 8.80552122]
])

colors = ['red', 'green', 'white', 'm']

totals = np.sum(data, axis=0)

percentages = np.round((data / totals) * 100).astype(int).astype(str)

labels_matrix = np.char.add(percentages, '%')

df = pd.DataFrame(data.T, index=people)

fig, ax = plt.subplots(figsize=(8, 6))

df.plot(kind='barh', stacked=True, color=colors, edgecolor='black', ax=ax, legend=False, width=0.8)

[ax.bar_label(container, labels=labels, label_type='center', fontsize=10) 
 for container, labels in zip(ax.containers, labels_matrix)]

ax.set_xlabel('Scores')
ax.set_xlim(0, 40)

plt.tight_layout()
plt.show()
```

### 17. Write a Python program to add textures (black and white) to bars and wedges.

```python
patterns = ('|', '/', '\\', '+', '-', '.', '*', 'x', 'o', 'O')

density = 2

fig, ax = plt.subplots(figsize=(8, 6))

for i, pattern in enumerate(patterns):
    dense_pattern = pattern * density 
    ax.bar(i + 0.5, 3, width=0.8, color='white', edgecolor='black', hatch=dense_pattern)
    
ax.set_xlim(0, 10)
ax.set_ylim(0.0, 3.0)

plt.show()
```
