```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv("alphabet_stock_data.csv")
df.head()
```

## 1. Write a Pandas program to create a line plot of the historical stock prices of Alphabet Inc. between two specific dates.

```python
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)

start_date = '2020-04-01'
end_date = '2020-04-30'
df_filtered = df.loc[start_date:end_date]
```

```python
plt.figure(figsize=(10, 4))
df_filtered['Close'].plot(kind='line', title='Alphabet Inc. Historical Stock Prices (Close)')
plt.ylabel('Price')
plt.show()
```

## 12. Write a Pandas program to create a plot of stock price and trading volume ofAlphabet Inc. between two specific dates.

## 13. Write a Pandas program to create a plot of Open, High, Low, Close, AdjustedClosing prices and Volume of Alphabet Inc. between two specific dates.

## 14. Write a Pandas program to create a plot of adjusted closing prices, thirty days and forty days simple moving average of Alphabet Inc. between two specificdates.

```python
df['SMA30'] = df['Adj Close'].rolling(window=30).mean()
df['SMA40'] = df['Adj Close'].rolling(window=40).mean()
df.loc[start_date:end_date][['Adj Close', 'SMA30', 'SMA40']].plot(figsize=(10, 5), title='Adj Close, SMA 30 & SMA 40')
plt.ylabel('Price')
plt.show()
```

## 15. Write a Pandas program to create a plot of adjusted closing prices, 30 days simple moving average and exponential moving average of Alphabet Inc. between two specific dates.

```python
df['EMA30'] = df['Adj Close'].ewm(span=30, adjust=False).mean()
df.loc[start_date:end_date][['Adj Close', 'SMA30', 'EMA30']].plot(figsize=(10, 5), title='Adj Close, SMA 30 & EMA 30')
plt.ylabel('Price')
plt.show()
```

## 16. Write a Pandas program to create a scatter plot of the trading volume/stock prices of Alphabet Inc. stock between two specific dates.

## 17. Write a Pandas program to create a plot to visualize daily percentage returns of Alphabet Inc. stock price between two specific dates.


```python
df['Daily Return'] = df['Close'].pct_change() * 100
df.loc[start_date:end_date]['Daily Return'].plot(figsize=(10, 4), marker='o', color='purple', title='Daily Percentage Returns')
plt.ylabel('Percentage %')
plt.show()
```

## 18. Write a Pandas program to plot the volatility over a period of time of Alphabet Inc. stock price between two specific dates.

```python
df['Volatility'] = df['Daily Return'].rolling(window=7).std()
df.loc[start_date:end_date]['Volatility'].plot(figsize=(10, 4), color='red', title='7-Day Rolling Volatility')
plt.ylabel('Volatility')
plt.show()
```

## 19. Write a Pandas program to create a histogram to visualize daily return distribution of Alphabet Inc. stock price between two specific dates.

```python
df.loc[start_date:end_date]['Daily Return'].plot(kind='hist', bins=15, color='teal', edgecolor='black', title='Daily Return Distribution')
plt.xlabel('Daily Return (%)')
plt.show()
```
