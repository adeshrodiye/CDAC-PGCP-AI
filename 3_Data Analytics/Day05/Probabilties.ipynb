# Joint Probability
- when event A and B are independent of each other - $$P(A \text{ and } B) = P(A \cap B) = P(A) \times P(B)$$
- EVENT A : Patient is diabetic
- EVENT B : Patient has heart disease
- P(diabetic and hear disease oatient = P(diabetic Pateint) x P(heart disease Patient)
- Both Event are Independent 

# Conditional Probability 
- When one is alredy happened and another one is about to happen P(B/A) = Probability of event B happening, Given That event A has happened
- $$P(B \mid A) = \frac{P(A \cap B)}{P(A)}$$
- $P(B \mid A)$: The conditional probability of event $B$ given event $A$.
- $P(A \cap B)$ (or joint probability): The probability that both events $A$ and $B$ happen together.
- $P(A)$: The marginal probability of event $A$ happening on its own.
- P(Patient has heart disease guven that patient has diabetic)
- First consider onlyy diabetic petients and within them look for pateints with heart disease

```python
df = pd.read_csv("titanic.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df = df[['gender','survived']]
df.head()
```

```python
df = df.dropna()
```

```python
total = len(df)
total
```

```python
# Marginal Probability of gender
total_female = len(df[df['gender']=='female'])
total_female
```

```python
total_male = len(df[df['gender']=='male'])
total_male
```

```python
p_female = total_female/total
p_female
```

```python
p_male = total_male/total
p_male
```

```python
# Marginal prob of Survival

total_survived = len(df[df['survived']==1])
total_survived
```

```python
total_not_survived = len(df[df['survived']==0])
total_not_survived
```

```python
p_survived = total_survived/total
p_survived
```

```python
p_not_survived = total_not_survived/total
p_not_survived
```

```python
print(f"P(male) : {p_male:.3f}")
print(f"P(female) : {p_female:.3f}")
print(f"P(survived=1) : {p_survived:.3f}")
print(f"P(survived=0) : {p_not_survived:.3f}")
```

# Joint probabilities

```python
total_survived_male = len(df[(df['survived']==1)& (df['gender']=='male')])
total_survived_male
```

```python
total_survived_female = len(df[(df['survived']==1)& (df['gender']=='female')])
total_survived_female
```

```python
total_not_survived_female = len(df[(df['survived']==0)& (df['gender']=='female')])
total_not_survived_female
```

```python
p_female_survived = total_survived_female/total
p_female_survived
```

```python
p_male_survived = total_survived_male/total
p_male_survived
```

```python
p_female_not_survived = total_not_survived_female/total
p_female_not_survived
```

```python
p_male_not_survived = total_not_survived_male/total
p_male_not_survived
```

```python
print(f"P(female,survived=1) : {p_female_survived :.3f}")
print(f"P(female,survived=0) : {p_female_not_survived:.3f}")
print(f"P(male,survived=1) : {p_male_survived:.3f}")
print(f"P(male,survived=0) : {p_male_not_survived:.3f}")
```
