```python
import pandas as pd
from scipy import stats
```

```python
df = pd.read_csv("tips.csv")
df.head()
```

```python
our_column = df["tip"]
our_column= our_column.sample(n=30, replace=False)

meu = 3

t_statistic, p_value = stats.ttest_1samp(our_column,meu)
print(p_value)
```

```python
print("P-value : ", p_value)
alpha =0.5
if p_value < alpha:
    print("Reject H0")
else:
    print("Fails to Reject H0")
```

## One Tailed One Sample T-Test

```python
our_col = df["tip"].sample(n=30,replace=False)
```

```python
meu = 3
result = stats.ttest_1samp(our_col, meu, alternative='greater') # Greater means H0 <= and H1 is > (no need to define meu < or >
print("T-statistic:",result.statistic)
print("P_value (one-tailed) : ", result.pvalue)
```

```python
alpha = 0.05
if result.pvalue < alpha:
    print("Reject H0 (means is significantly greater than meu")
else:
    print("Do not reject H0")
```

## Ex: Use Loan datset. run 2 tailed and one tailed one sample t-test
## 2-tailed -> h0: Avg income = 50 , H1 : Average income <> 50
## 1 tailed -> H0: Avg income >= 50, H1 : Avg income <50

```python
df = pd.read_csv("loan.csv")
df.head()
```

### 2 tailed

```python
our_col = df["income"].sample(n=30,replace=False)
meu = 50

result = stats.ttest_1samp(our_col, meu, alternative="two-sided") # less means H0 >= and H1 is <
print("T-statistic:",result.statistic)
print("P_value (one-tailed) : ", result.pvalue)

alpha = 0.05
if result.pvalue < alpha:
    print("Reject H0 (means is significantly greater than meu")
else:
    print("Do not reject H0")
```

### one tailed

```python
our_col = df["income"].sample(n=30,replace=False)
meu = 50
```

```python
result = stats.ttest_1samp(our_col, meu, alternative='less') # less means H0 >= and H1 is <
print("T-statistic:",result.statistic)
print("P_value (one-tailed) : ", result.pvalue)
```

```python
alpha = 0.05
if result.pvalue < alpha:
    print("Reject H0 (means is significantly greater than meu")
else:
    print("Do not reject H0")
```
