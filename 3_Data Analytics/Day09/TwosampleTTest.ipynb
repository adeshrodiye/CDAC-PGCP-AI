```python
import pandas as pd
from scipy import stats
```

```python
data = pd.read_csv("../Files/tips.csv")
male_tips = data[data["gender"] == "Male"]["tip"]
female_tips = data[data["gender"] == "Female"]["tip"]
```

```python
print(f"Male mean tip : {male_tips.mean()}")
print(f"Female mean tip : {female_tips.mean()}")

```

```python
# perform 2 sample t test

t_statistic , p_value = stats.ttest_ind(male_tips, female_tips)
print(f"t-stat : {t_statistic}")
print(f"p_value : {p_value}" )
```

```python
alpha = 0.05
if p_value < alpha:
    print(f"There is significant difference in spending score between Male and Female groups.")
else:
    print(f"There is no significant difference in spending score between Male and Female groups.")
```

### Exercise : Is there a statistically significant difference in the average runs scored by sachine tendulkar when playing against australia versus pakistan

```python
df = pd.read_csv("../Files/Sachin_Tendulkar.csv")
df
```

```python
runs_aus = df[df["Versus"] == "Australia" ]["Runs"]
runs_pak = df[df["Versus"] == "Pakistan"]["Runs"]
```

```python
print(f"Mean agaisnt aus : {runs_aus.mean()}")
print(f"Mean against pak : {runs_pak.mean()}")
```

```python
t_statistic , p_value = stats.ttest_ind(runs_aus, runs_pak)
print(f"t-stat : {t_statistic}")
print(f"p_value : {p_value}" )
```

```python
alpha = 0.05
if p_value < alpha:
    print(f"There is significant difference in average runs scored by sachin when playing against aus and pak.")
else:
    print(f"There is no significant difference in average runs scored by sachin when playing against aus and pak.")
```
