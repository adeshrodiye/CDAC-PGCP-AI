```python
import pandas as pd
from scipy import stats
```

```python

df = pd.read_csv("../Files/cust_seg.csv")
df
```

```python
usage_data = df[["pre_usage","Post_usage_1month"]]
usage_data = usage_data.sample(n=30)
```

```python
tstat, pval = stats.ttest_rel(usage_data["pre_usage"],usage_data["Post_usage_1month"])
print(f"tstat : {tstat}")
print(f"pval : {pval}")
```

```python
alpha = 0.05
if pval < alpha:
    print(f"Reject null hypothesis")
else:
    print("Fail to reject null hypothesis")
```

### Exercise : On the hypothermia dataset , check whether the urgent treatment is effective (Consider columns t.1 and t.2)

```python
df = pd.read_csv("../Files/Hypothermia.csv")
df
```

```python
usage_data = df[["t.1","t.2"]]
usage_data = usage_data.sample(n=30)
```

```python
tstat, pval = stats.ttest_rel(usage_data["t.1"],usage_data["t.2"])
print(f"tstat : {tstat}")
print(f"pval : {pval}")
```

```python
alpha = 0.05
if pval < alpha:
    print(f"Reject null hypothesis")
else:
    print("Fail to reject null hypothesis")
```
