```python
import pandas as pd
from scipy import stats
```

```python
df = pd.read_csv("../Files/tips.csv")
our_column = df["tip"]
our_column= our_column.sample(n=30, replace=False)

meu = 3

t_statistic, p_value = stats.ttest_1samp(our_column,meu)
print(p_value)
```

```python
print("P-value : ", p_value)
alpha =0.5
if p_value < alpha:
    print("Reject H0")
else:
    print("Fails to Reject H0")
```

### One tailes One sample T-test

```python
our_column = df["tip"].sample(n=30, replace=False)

meu = 3

```

```python
result = stats.ttest_1samp(our_column,meu, alternative="greater")

print("T-Statistic : " , result.statistic)

```

```python
print(f"p_value (one-tailed) : {result.pvalue}" )
```

```python
alpha = 0.05
if result.pvalue < alpha:
    print(f"Reject H0 (mean is significantly greater than {meu})")
else:
    print(f"Do not reject H0")
```

### Use loan dataset. Run two-tailed and one-tailed one sample t-test
 - Two-Tailed -> H0 : Average income = 50,H1: Average income != 50
 - one-tailed -> h1 : Average income >= 50, H1 : Average income < 50

```python
df = pd.read_csv("../Files/loan.csv")
df
```

```python
our_column = df["income"].sample(n=30, replace=False)

meu = 50

result = stats.ttest_1samp(our_column,popmean=meu, alternative="two-sided")
print(f"T-statistics : {result.statistic}")
print(f"P-Value : {result.pvalue}")
```

```python
alpha = 0.05
if result.pvalue < alpha:
    print(f"Reject H0 (mean is significantly greater than {meu})")
else:
    print(f"Do not reject H0")
```

```python
## One-tailed

our_column = df["income"].sample(n=30, replace=False)

meu = 50

result = stats.ttest_1samp(our_column,popmean=meu, alternative="less")
print(f"T-statistics : {result.statistic}")
print(f"P-Value : {result.pvalue}")
```

```python
alpha = 0.05
if result.pvalue < alpha:
    print(f"Reject H0 (mean is significantly greater than {meu})")
else:
    print(f"Do not reject H0")
```
