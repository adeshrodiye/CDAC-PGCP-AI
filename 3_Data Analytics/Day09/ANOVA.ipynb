```python
from scipy.stats import f_oneway
from pandas import read_csv
```

```python
df = read_csv("../Files/studentsPerformance.csv")
```

```python
df
```

```python
q1 = df["math score"][df["parental level of education"] == "some high school"]
q2 = df["math score"][df["parental level of education"] == "high school"]
q3 = df["math score"][df["parental level of education"] == "some college"]
q4 = df["math score"][df["parental level of education"] == "associate\'s degree"]
q5 = df["math score"][df["parental level of education"] == "Bachelor\'s degree"]
q6 = df["math score"][df["parental level of education"] == "master\'s degree"]
```

```python
q3
```

```python
fstat,pval = f_oneway(q1,q2,q3,q4,q5,q6)
```

```python
alpha = 0.05
if pval < alpha:
    print("Reject H0")
else:
    print("Do not reject H0")
```

### Exercise : is there a statistically significant difference between how people tip, whose total bill amounts are < $5 , $5-10, $10-15 , >$15 in the tips dataset ? 

```python
df = read_csv("../Files/tips.csv")
df
```

```python
q1 = df["tip"][df["total_bill"] < 5]
q2 = df["tip"][(df["total_bill"] >= 5) & (df["total_bill"] < 10)]
q3 = df["tip"][(df["total_bill"] >= 10) & (df["total_bill"] <= 15)]
q4 = df["tip"][df["total_bill"] > 15]
```

```python
fstat,pval = f_oneway(q1,q2,q3,q4)
print(f"fstat : {fstat}")
print(f"pvalue : {pval}")
```

```python
alpha = 0.05
if pval < alpha:
    print("Reject H0")
else:
    print("Do not reject H0")
```
